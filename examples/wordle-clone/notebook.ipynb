## Core scoring logic

The classic Wordle logic bug is **repeated letters**: naively checking "is this guess-letter anywhere in the
target?" over-counts when the guess or target repeats a letter. The real rules are a two-pass algorithm — mark exact
position matches (green) first, then only flag a *remaining, unconsumed* copy of a letter as yellow. This is copied
verbatim from `wordle.py` in the real example.

```python
from collections import Counter

WORD_LENGTH = 5
GREEN, YELLOW, GRAY = "G", "Y", "X"


def score_guess(guess: str, target: str) -> list[str]:
    guess, target = guess.upper(), target.upper()
    marks = [GRAY] * WORD_LENGTH

    remaining = Counter()
    for i, (g, t) in enumerate(zip(guess, target)):
        if g == t:
            marks[i] = GREEN
        else:
            remaining[t] += 1

    for i, g in enumerate(guess):
        if marks[i] == GREEN:
            continue
        if remaining[g] > 0:
            marks[i] = YELLOW
            remaining[g] -= 1

    return marks


# Sanity check against a classic repeated-letter case: target has one E,
# guess has two.
print(score_guess("SPEED", "ERASE"))  # -> ['Y', 'X', 'Y', 'Y', 'X']
```

## Stats tracking

Same shape as `stats.py` in the real example: games played, win rate, current/max streak, and a guess-count
histogram, persisted to a local JSON file.

```python
import json
from pathlib import Path

STATS_FILE = Path("stats.json")
MAX_GUESSES = 6
DEFAULT_STATS = {
    "played": 0,
    "wins": 0,
    "current_streak": 0,
    "max_streak": 0,
    "guess_distribution": {str(n): 0 for n in range(1, MAX_GUESSES + 1)},
}


def load_stats():
    if not STATS_FILE.exists():
        return json.loads(json.dumps(DEFAULT_STATS))
    return json.loads(STATS_FILE.read_text())


def save_stats(stats):
    STATS_FILE.write_text(json.dumps(stats, indent=2))


def record_result(stats, won, guesses_used):
    stats["played"] += 1
    if won:
        stats["wins"] += 1
        stats["current_streak"] += 1
        stats["max_streak"] = max(stats["max_streak"], stats["current_streak"])
        stats["guess_distribution"][str(guesses_used)] += 1
    else:
        stats["current_streak"] = 0
    return stats


def print_stats(stats):
    played = stats["played"]
    rate = stats["wins"] / played if played else 0.0
    print(f"Played: {played}  Win rate: {rate:.0%}  "
          f"Streak: {stats['current_streak']} (max {stats['max_streak']})")
    print("Guess distribution:", stats["guess_distribution"])
```

## Play a round

Type a 5-letter word and press Enter. Feedback is printed as one letter per guessed letter: `G` = green (right spot),
`Y` = yellow (right letter, wrong spot), `X` = gray (not in the word, or no more unused copies of it). Guesses that
aren't 5 letters or aren't in `WORD_SET` are rejected with a prompt to try again — same dictionary-validation rule as
the real example.

```python
import random

def play_round():
    target = random.choice(WORDS)
    print(f"Guess the {WORD_LENGTH}-letter word. You have {MAX_GUESSES} tries.\n")

    for attempt in range(1, MAX_GUESSES + 1):
        guess = input(f"Guess {attempt}/{MAX_GUESSES}: ").strip().upper()
        if len(guess) != WORD_LENGTH or not guess.isalpha():
            print(f"  Please enter exactly {WORD_LENGTH} letters."); continue
        if guess not in WORD_SET:
            print(f"  '{guess}' isn't in the word list, try a real word."); continue

        marks = score_guess(guess, target)
        print("  " + " ".join(f"{l}:{m}" for l, m in zip(guess, marks)))

        if all(m == GREEN for m in marks):
            print(f"\nYou got it in {attempt}!")
            return True, attempt

    print(f"\nOut of guesses. The word was {target}.")
    return False, MAX_GUESSES


stats = load_stats()
won, attempts = play_round()
stats = record_result(stats, won, attempts)
save_stats(stats)
print_stats(stats)
```
