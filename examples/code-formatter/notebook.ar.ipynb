**👟 تلميح البداية :** `open(...).read().splitlines()` — أسطر دون `\n` البالية، لذا كل مدخل محتوى خالص.

### 1.2 ابحث عن العلامات القذرة

**👟 تلميح البداية :** ثلاثة مسوحات: مسافات بائية، تعليقات ملتصقة بالنص، ومسافات بادئة غير رباعية.

## الخطوة 2: شذّب وأصلح التعليقات

التنظيف الآمن: أزل المسافات البالية، ثم ضع مسافة بعد كل `#`.

### 2.1 الأدوات الطبيعية لكل سطر

**👟 تلميح البداية :** دالة `fix_line` تفرّغ الطرف ثم تصلح جزء التعليق بـ `partition("#")`.


````python
# code_formatter.py (continued)
def fix_line(ln):
    fixed = ln.rstrip()
    if "#" not in fixed or fixed.startswith("#!"):
        return fixed
    pre, _, comment = fixed.partition("#")
    comment = comment.strip()
    if comment == "":
        return pre.rstrip()
    if pre.strip() == "":
        return "# " + comment
    return pre.rstrip() + "  # " + comment
````

تقسيم `partition("#")` يُبقي الجانب الأيسر (الكود) منفصلًا عن التعليق، لذا كل جانب يُطبَّع بشكل مستقل. تعليق في العمود 0 ← `# sum module`؛ تعليق داخل سطر ← كود، مسافتان، `# comment`. يُترك `#!` (رأس سكربت shebang) وحده — له اصطلاحه الخاص.

**🎯 الناتج المتوقع :** دالة، لا مخرجات بعد — لكن فكّر ماذا تفعل بالسطر 10: `   return msg   #too much indent` ← `   return msg  # too much indent`.

### 2.2 طبّقها على الملف كاملًا

**👟 تلميح البداية :** اربط `fix_line` فوق كل الأسطر واطبع النتيجة.


````python
# code_formatter.py (continued)
fixed = [fix_line(ln) for ln in lines]
for i, ln in enumerate(fixed, 1):
    print(f"{i:>2} |{ln}|")
````

```bash
10 |   return msg  # too much indent|
```


**🩹 إذا لم يعمل :** إذا أصبح السطر 1 `  # sum module`، فلن يعمل فرع العمود-0 (`pre.strip() == ""`) — تحقق أنك *قسّمت* قبل فحص `pre`. إذا ما زال `#` يلامس النص، فقد تُخطّي `comment.strip()` ولم تُدرج المسافة.

### 2.3 تحقّق من التنظيف

**✅ قائمة التحقق**

- ✅ اختفت المسافات البالية من 1 و2 و5 و7.
- ✅ `#sum module` ← `# sum module`؛ `#add two numbers` ← `# add two numbers`؛ `#too much indent` ← `# too much indent`.
- ✅ يحتفظ الكود والتعليق بمسافتين بالضبط بينهما — عقد `pre.rstrip() + "  # "`.

**🤔 سؤال (أسئلة) سقراطي(ة)**

- تعامل `fix_line` الجانب *الأيسر* ككود. ماذا سيحدث لسلسلة Python *نصية* تحوي `#` (`s = "color #ff00aa"`)? هل يستحق منسّق مدرك للسلاسل التعقيد لأداة أولى — وماذا يقول ذلك عن حدود "المجموعة الفرعية الآمنة"؟
- يُستثنى `#!` بحالة خاصة. تعليقات سطر باش (`#!`، `##`)، والسلسلة docstrings (`"""`)، والسلاسل النصية المضمّنة كلها تفرط في تحميل `#`. أي قاعدة إبهام واحدة تُبقي منسّقًا مبتدئًا بعيدًا عن إفساد الملفات الصالحة؟

## الخطوة 3: طيّ الأسطر الفارغة المتتالية وفرض سطر نهائي

الأسطر الفضائية-فقط والأسطر الفارغة المتكررة ضجيج تخطيطي. تشدّ الخطوة 3 منها.

### 3.1 سطر فارغ واحد في كل مرة

**👟 تلميح البداية :** امشِ فوق الأسطر المُصلحة، متجاهلًا أي سطر فارغ (`""` أو مسافات-فقط) يلي سطرًا فارغًا آخر مباشرة.


````python
# code_formatter.py (continued)
def collapse_blanks(lines):
    out = []
    for ln in lines:
        blank = ln.strip() == ""
        if blank and out and out[-1].strip() == "":
            continue
        out.append(ln)
    return out

collapsed = collapse_blanks(fixed)
print("lines after collapse:", len(collapsed))
for i, ln in enumerate(collapsed, 1):
    print(f"{i:>2} |{ln}|")
````

`strip() == ""` يسمّي سطرًا فارغًا *سواء* كان سطرًا خاويًا حقًا أو سطرًا مسافات-فقط (`   `) — كلاهما تخطيط، ولا يحمل أيٌّ منهما محتوى. الحارس `out and out[-1].strip() == ""` يُبقي فقط *الأول* من المتتالية، لذا يتطابق سطران فارغان+ إلى سطر واحد في كل مكان في مسح واحد.

**🎯 الناتج المتوقع :**

```bash
 5 |        return a + b|

 8 |  msg = "hello " + name|
 9 |   return msg  # too much indent|
```


**🩹 إذا لم يعمل :** إذا ما زال السطر 7 يُطبع فارغًا، فإن سطر المسافات-فقط لم يُفرَّغ بـ `strip() == ""` — لقد فُرّغ، ما لم يحمل السطر حروفًا غير مرئية غير مسافات. إذا تركت *متتالية من ثلاثة* سطرين فارغين، فحص الحارس السطر الخام `ln` بدلًا من السطر المُضاف أخيرًا.

### 3.2 السطر النهائي

**👟 تلميح البداية :** أعد الوصل بـ `"\n"` وأنهِ النص دائمًا بـ `"\n"`.


````python
# code_formatter.py (continued)
def build_text(lines):
    return "\n".join(lines) + "\n"

text = build_text(collapsed)
print("ends with newline:", text.endswith("\n"))
print("input bytes:", len(open("messy.py").read().encode()),
      "output bytes:", len(text.encode()))
````

يجب أن ينتهي آخر سطر في ملف بسطر جديد — اصطلاح POSIX، والشيء الذي يضمنه `join + "\n"` حتى عندما نسي المصدر ذلك. عدّادات البايت فحص صحة سريع: التنظيف *يُصغّر* الملف (177 ← 166 بايتًا) لأن المسافات القمامة بايتات حقيقية.

**🎯 الناتج المتوقع :**


```bash

```


**🩹 إذا لم يعمل :** إذا كانت `ends with newline: False`، فقد وقع `+ "\n"` قبل `join`. إذا كانت بايتات المخرجات *أكبر*، فقد أضافت تطبيع التعليقات مسافات أسرع مما أزال تشذيب البالية — قِس بأمانة، هذا حكم الأداة.

### 3.3 تحقّق من الطيّ

**✅ قائمة التحقق**

- ✅ 10 أسطر ← 9: اندمج سطر `   ` الفضائي-فقط مع السطر الفارغ فوقه.

**🤔 سؤال (أسئلة) سقراطي(ة)**

- يضيف `build_text` `\n` واحدة للملف كاملًا. لماذا هي *السطر الجديد الوحيد* الذي يحتاج ذلك العدد — وماذا سيفعل `"\n".join(lines)` *دون* السطر الجديد البالي بـ `splitlines()` في القراءة التالية؟

## الخطوة 4: تشخيص المسافات البادئة

المسافات البادئة في Python دلالات، لذا المنسّق *يشخّص* بدلًا من التخمين.

### 4.1 أصدر تحذيرات

**👟 تلميح البداية :** أعد تشغيل مسح عدّ البادئات واطبع كل سطر غير مضاعف للعدد 4 مع عدده الحالي من المسافات.


````python
# code_formatter.py (continued)
print("INDENT WARNINGS")
for i, ln in enumerate(collapsed, 1):
    body = ln.lstrip(" ")
    if body and not body.startswith("#"):
        lead = len(ln) - len(ln.lstrip(" "))
        if lead % 4 != 0:
            print(f"  line {i}: {lead} spaces (should be a multiple of 4)")
````

يرفض المنسّق تخمين الإصلاح — مسافتان في السطر 8 و3 في السطر 9 غامضتان (الاثنتان تتبعان `def`، لكن الأداة لا تعرف السياق)، لذا تعرضهما لعين المطوّر.

**🎯 الناتج المتوقع :**


```bash

  line 9: 3 spaces (should be a multiple of 4)
```


**🩹 إذا لم يعمل :** إذا سمّت التحذيرات أسطرًا مختلفة، فللقائمة المطوية مواضع مختلفة عن `messy.py` — التقرير عن *النص الحالي*. إذا لم يُطبع شيء، فإن `lstrip(" ")` على سطر مُبطّن بتبويب يخفي البادئة (انظر سقراطي أدناه).

### 4.2 عقد التبويب

**👟 تلميح البداية :** حوّل أي متتاليات تبويب موجودة إلى كتل رباعية وسجّل هل وُجد أيٌّ منها.


````python
# code_formatter.py (continued)
has_tabs = any("\t" in ln for ln in collapsed)
print("tabs found in source:", has_tabs)
````

`\t` محظور في العيّنة (وعادة في مصدر Python حسب PEP 8). الفحص هو `any(...)` واحد فوق الأسطر؛ إذا وُجد، كان `.expandtabs(4)` سيعيد كتابتها — لكن بما أن `messy.py` لا يحوي أيًّا، فالجواب المطبوع `False`، وتبقى قصة التبويب عقدًا موثقًا بدلًا من تحويل خفي.

**🎯 الناتج المتوقع :** `tabs found in source: False`

**🩹 إذا لم يعمل :** إذا طُبع `True`، فقد اكتسبت نسختك تبويبًا في مكان ما — قرّر: أبقِه تشخيصيًا (أبلغ عن السطر) أو وسّعه بـ `.expandtabs(4)`، وعدّل محاسبة المسافات لاحقًا.

### 4.3 تحقّق من التشخيص

**✅ قائمة التحقق**

- ✅ تسمّي التحذيرات السطرين 8 (مسافتان) و9 (3 مسافات) — كلاهما بيانان، لا تعليقان.
- ✅ `tabs found in source: False`.

**🤔 سؤال (أسئلة) سقراطي(ة)**

- `def` في العمود 0، وجسمه في 4، والأجسام المتداخلة في 8. بهذه الحقائق الثلاث، هل توجد *أي* قاعدة لا لبس فيها "لإصلاح" المسافات البادئة لسطر مُبطَّن — أم التحذير هو المنتج الصحيح هنا؟

### 5.2 الموجّه (المُوزِّع)

**👟 تلميح البداية :** اقرأ `sys.argv[1]` كاسم الملف، ونسّقه، واطبع الملخص قبل/بعد.


````python
# code_formatter.py (continued)
def format_file(path):
    lines = open(path).read().splitlines()
    fixed = [fix_line(ln) for ln in lines]
    collapsed = collapse_blanks(fixed)
    text = build_text(collapsed)
    with open("formatted.py", "w") as f:
        f.write(text)
    trailing = [i for i, ln in enumerate(lines, 1) if ln != ln.rstrip()]
    print(f"{path}: {len(lines)} -> {len(collapsed)} lines; "
          f"{len(trailing)} trailing-whitespace fixes; "
          f"see formatted.py")

if __name__ == "__main__":
    format_file(sys.argv[1])
````

خط الأنابيب كاملًا — قراءة، إصلاح، طيّ، وصل، كتابة، تلخيص — أصبح الآن *دالة واحدة* لمسار ملف. `sys.argv[1]` يحوّله إلى CLI: اكتب `python3 code_formatter.py messy.py` والأداة تحرّر من سطر الأوامر.

## ⚠️ المآزق الشائعة

- **`splitlines` مقابل `read().split("\n")`.** يتجاهل `splitlines()` العنصر الفارغ الأخير الذي يُنتجه تقسيم `"\n"` ساذج — لتنتهي بسطر فارغ زائف.
- **`lstrip()` يجرّد التبويبات أيضًا.** عدّ المسافات البادئة بـ `len(ln) - len(ln.lstrip())` يعدّ المسافات *والتبويبات* كحرف واحد؛ استخدم `lstrip(" ")` أو مسحًا مدركًا للتبويبات. (يفحص هذا المشروع التبويبات بشكل منفصل.)
- **طيّ الفراغات على القائمة الخاطئة.** الطيّ قبل التشذيب يعني أن السطر الفضائي-فقط (`   `) يتصرف كـ *محتوى* ولا يندمج أبدًا مع الفراغ فوقه. الترتيب: شذّب ← أصلح ← اطوِ.
- **`partition` مقابل `split`.** يُبقي `partition("#")` القطع الثلاث كلها (pre, "#", post)؛ كان `split("#")` ليعامل تعليقًا يحوي `#` بشكل خاطئ أو يدمر الحد عند أول فاصل.
- **تباعد التعليقات على السلاسل النصية.** `s = "#ff00aa"` يحوي `#` *داخل سلسلة نصية حرفية* — منسّق مسافات-فقط يعيد كتابتها بسعادة. حدود "المجموعة الفرعية الآمنة" درعك؛ وثّقها.
- **الكتابة فوق العيّنة.** إعادة كتابة `messy.py` تدمر الشيء الذي تقيسه. أخرج إلى `formatted.py`؛ أبقِ المدخل للقراءة فقط.
