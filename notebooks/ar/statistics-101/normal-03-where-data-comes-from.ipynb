## ماذا يعني

لكل مجموعة بيانات **مصدر** (provenance): من أو ما أنتج كل صف، ومتى، وفي أي ظروف. ويحدد المصدر ما تستطيع البيانات الإجابة عنه، ولا يستطيع أي قدر من التحليل إضافة معلومة لم تلتقطها خطوة الجمع أصلًا.

أهم المصادر التي ستلتقيها:

| المصدر | ما يقدمه | ما يفتقده عادةً |
|---|---|---|
| **استطلاع** | آراءً ووقائع يبلّغ عنها الشخص | الصدق، والأشخاص الذين لم يجيبوا |
| **مستشعر أو سجل** | توقيتات وأعدادًا دقيقة | السياق — لماذا حدثت القيمة |
| **سجلات إدارية** | تغطية كاملة لعملية ما | أي شيء خارج تلك العملية |
| **تجربة** | دليلًا على السببية | الواقعية — السياق مصطنع |
| **تحليلات الويب** | سلوكًا على نطاق واسع | الهوية، وسبب النقرة |

**تحيز المعاينة** (sampling bias) هو ما يحدث عندما تختلف الحالات المرصودة اختلافًا منهجيًا عن المجتمع. مثالان يجعلانه ملموسًا:

- استطلاع رضا إلكتروني لا يجيب عنه إلا من لم يغادر الموقع — فالمغادرون هم بالضبط غير الراضين، وهم مفقودون؛
- دراسة لاستخدام تطبيق من السجلات، حيث لا يظهر فيها أصلًا من لم يثبّت التطبيق قط.

في كلتا الحالتين تكون العينة كبيرة ونظيفة ومقيسة بدقة، وتجيب عن سؤال مختلف عن السؤال المطروح.

## لماذا يضيف شيئًا

يضع سقفًا لما يمكن لأي تحليل لاحق أن يدّعيه. تخبرك فترة الثقة بمدى دقة قياسك لـ*مجتمع العينة*؛ ولا تستطيع إخبارك هل ذلك المجتمع هو الذي يهمك. ومعرفة المصدر تخبرك أي تحفظ تلحقه قبل أن تحسب متوسطًا أصلًا.

التمييز الآخر المهم هو **الرصدي مقابل التجريبي**:

- في دراسة **رصدية** تراقب ما يحدث: بعض الناس اختاروا ممارسة الرياضة وبعضهم لا، وتقارن بينهم؛
- في **تجربة** تسند العلاج: أنت تقرر من يمارس الرياضة ومن لا يمارسها.

الثانية وحدها تستطيع إسناد ادعاء سببي بنفسها، لأن الإسناد العشوائي يقطع الصلة بين العلاج وكل شيء آخر عن الشخص.

## مثال محسوب

تريد شركة معرفة هل تزيد مسار دفع جديد من المشتريات.

- **إذا قارنوا العملاء الذين صادف أن رأوا المسار الجديد بمن لم يره**، فلديهم مقارنة رصدية. تختلف المجموعتان في أمور مهمة: فقد أُطلق المسار الجديد على جهاز معين أو لحسابات أحدث، وتلك الفروق متشابكة مع النتيجة. وأي فرق هو مزيج من المسار ومن طريقة الإطلاق.
- **إذا أسندوا الزوار عشوائيًا إلى المسار القديم أو الجديد**، فلديهم تجربة. تختلف المجموعتان بالمصادفة وحدها، فالفرق في معدل الشراء يُنسب إلى المسار — ضمن نطاق خطأ المعاينة.

الحساب بعد ذلك متطابق. والفرق كله في خطوة الجمع، وهو الفرق بين تلميح ودليل.

## الأسئلة الأربعة قبل أي تحليل

1. **ما الذي يمثله الصف الواحد؟** توصيل، أم شخص، أم جلسة، أم يوم؟ الخطأ هنا يجعل كل تعداد خاطئًا.
2. **من الغائب عن هذه البيانات؟** غير المستجيبين، ومن لم يثبّتوا التطبيق، والمعاملات الفاشلة قبل التسجيل.
3. **كيف اختيرت الحالات؟** عشوائيًا، أم بالسهولة، أم بحسب من كان مسجلًا؟
4. **هل أُسند شيء، أم رصد فقط؟** هذا يحدد هل ادعاء السببية متاح أصلًا.

## متى تستخدمه

كل مرة تفتح مجموعة بيانات غير مألوفة، وكلما كانت نتيجة على وشك أن ترشد قرارًا. فحص المصدر رخيص، وهو الخطوة الوحيدة التي لا يمكن استدراكها لاحقًا.

## القيود

- **المصدر غير موثق غالبًا.** حينها تعيد بناءه، أو تعامل الاستنتاجات كمؤقتة.
- **حتى العينة النظيفة لا تصلح متغيرًا مفقودًا.** إذا لم يسجل الرضا للإلغاءات قط، فلا تحليل يستعيده.
- **العشوائية ليست أخلاقية أو ممكنة دائمًا.** حينها تكون الطرق الرصدية بتعديل حذر البديل الصادق، مع ذكر حدودها.

## أخطاء شائعة

- **الخلط بين عينة كبيرة وعينة ممثلة.** مليون صف متحيز يبقى متحيزًا.
- **افتراض أن السجلات كاملة.** تفشل التسجيلات، وتوجد مانعات الإعلانات، وحركة الروبوتات حقيقية.
- **قراءة سببية من مقارنة رصدية** لأن الحساب كان متطورًا.
- **تجاهل غير المستجيبين**، وهم عادةً المجموعة الأكثر إفادة في الاستطلاع.
