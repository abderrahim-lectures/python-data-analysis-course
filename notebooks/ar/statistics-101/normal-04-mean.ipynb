## ماذا يعني

**المتوسط** (mean) هو العدد الذي يجب أن تكون عليه كل قيمة لو تساوت جميعها وبقي المجموع كما هو. تحسبه بجمع كل شيء والقسمة على عدد القيم.

للقيم $x_1, x_2, \dots, x_n$:

$$
\bar{x} = \frac{x_1 + x_2 + \cdots + x_n}{n} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

اقرأ الطرف الأيمن كتعليمات: *اجمع قيم $x$، ثم وزّع المجموع بالتساوي على $n$ مواقع.* لا شيء مخفي في تلك الصيغة، وهذه قوتها.

## لماذا يضيف شيئًا

رقم واحد يلخّص أرقامًا كثيرة. بدلًا من أن تحمل في رأسك 200 سعر دراجة، تحمل رقمًا واحدًا يحفظ خاصية مهمة للقائمة كلها: مجموعها. إذا عرفت المتوسط والعدد، استطعت استعادة المجموع بدقة:

$$
\text{sum} = n \cdot \bar{x}.
$$

لا يحفظ أي مقياس مركز آخر المجموع. لهذا يظهر المتوسط في كل ما يتصل بالمال، من متوسط قيمة الطلب إلى الناتج المحلي للفرد: المجاميع تهم، والمتوسط هو المجموع مقسومًا على العدد.

## مثال محسوب يدويًا

خذ خمسة أزمنة توصيل بالدقائق:

$$
8,\; 12,\; 10,\; 15,\; 45
$$

الخطوة 1 — اجمعها:

$$
8 + 12 + 10 + 15 + 45 = 90
$$

الخطوة 2 — اقسم على العدد، $n = 5$:

$$
\bar{x} = \frac{90}{5} = 18
$$

متوسط زمن التوصيل 18 دقيقة. تحقق من المعنى: لو استغرقت التوصيلات الخمسة كلها 18 دقيقة، لبقي المجموع 90. هذا بالضبط ما يعد به المتوسط.

## الشيء نفسه في بايثون


```python
times = [8, 12, 10, 15, 45]

total = sum(times)
n = len(times)
print(total / n)                 # 18.0
```

الحساب اليدوي أولًا مهم، لأن نسخة المكتبة تخفي الخطوتين. تحتوي مكتبة بايثون القياسية على الحساب نفسه:


```python
from statistics import mean

print(mean([8, 12, 10, 15, 45]))  # 18
```

كلاهما صحيح. الصيغة الطويلة تستحق المعرفة لأنه عندما تبدو الإجابة خاطئة، تحتاج إلى معرفة أي جزء تفحصه.

## رؤيته هندسيًا

يمكن رسم كل قيمة نقطةً على خط، والمتوسط هو النقطة التي يتوازن عندها الخط، كأرجوحة بأثقال:

- القيمة البعيدة عن المتوسط تسحب بقوة أكبر، لأن مسافتها تُحتسب أكثر؛
- المسافات فوق المتوسط وتحته تتعادل تمامًا.

يمكنك سحب نقاط على خط أعداد ومراقبة حركة نقطة التوازن في [مستكشف المتوسط والوسيط من GeoGebra](https://www.geogebra.org/m/zsbf4jqm). لاحظ ما يحدث حين تسحب نقطة واحدة بعيدًا إلى اليمين: تتبعها نقطة التوازن.

## متى تستخدمه

- تهتم بـ**المجموع** لا بالفرد: متوسط الإنفاق لكل عميل، متوسط الدرجة، متوسط درجة الحرارة.
- تكون البيانات متماثلة تقريبًا، دون قيم متطرفة في أحد الطرفين.
- ستحسب به لاحقًا: المتوسطات تتجمع بطريقة لا تتجمع بها الوسيطات، ولهذا تستخدم كل صيغة تقريبًا لاحقًا في هذا المقرر المتوسط.

## القيود

- **تسحبه القيم الشاذة.** قيمة واحدة كبيرة جدًا أو صغيرة جدًا قد تبعده عن بقية البيانات.
- **يحتاج أرقامًا.** لا يوجد متوسط لـ"أحمر، أخضر، أزرق".
- **قد يكون قيمة لا يملكها أحد.** متوسط 2 و3 هو 2.5؛ وإذا كان المتغير لا يأخذ إلا أعدادًا صحيحة، فقد تكون 2.5 مستحيلة.
- **لا يقول شيئًا عن التشتت.** مجموعتا بيانات شديدتا الاختلاف في التباين قد تتشاركان المتوسط. الوحدة التالية تعالج ذلك.

## أخطاء شائعة

- **القسمة على العدد الخطأ.** خطأ الواحد بسبب إدراج صف عنوان أو خلية فارغة هو أشهر خطأ في المتوسط في العمل الحقيقي.
- **حساب متوسط المتوسطات.** متوسط ثلاثة متوسطات جماعية يساوي المتوسط الكلي فقط عندما تكون المجموعات بالحجم نفسه. وإلا وجب الترجيح بحجم المجموعة.
- **الإبلاغ بأرقام أكثر مما تحتمله البيانات.** إذا سُجلت الأزمنة إلى أقرب دقيقة، فأبلغ عن 18 دقيقة لا 18.0.
