## ماذا يعني

**المنوال** (mode) هو القيمة الأكثر تكرارًا. بخلاف المتوسط والوسيط لا يحتاج إلى أي حساب — إلى العد فحسب — وهذا يعني أنه مقياس المركز الوحيد الذي يعمل مع الأسماء والألوان والفئات.

$$
\text{mode} = \text{the value with the highest frequency.}
$$

قد تكون القائمة بلا منوال (كل قيمة تظهر مرة واحدة) أو بمنوالات عدة (قيمتان أو أكثر تتعادلان في أعلى تكرار). كلاهما طبيعي، وليس أي منهما خطأً.

## لماذا يضيف شيئًا

اسأل "ما متوسط لون السيارة؟" فيصبح المتوسط عديم المعنى والوسيط غير معرّف. يجيب المنوال: الأبيض، لأن الأبيض يبيع أكثر.

يخبرك المنوال أيضًا بشيء لا يستطيعه المتوسط والوسيط: أين تقع *الكتلة* العظمى من البيانات. إذا كان للتوزيع قمتان، يقع المتوسط بينهما، في وادٍ لا تكاد توجد فيه بيانات. أما المنوال فيشير إلى القمتين، ووجود منوالين تلميح قوي إلى أن مجموعة بياناتك في الحقيقة مجموعتان ملتصقتان.

## مثال محسوب يدويًا

أحجام أحذية بيعت في متجر في يوم واحد:

$$
38,\;39,\;39,\;40,\;40,\;40,\;41,\;42,\;42
$$

عُدّ كل قيمة: 38 مرة، 39 مرتين، 40 ثلاث مرات، 41 مرة، 42 مرتين. المنوال 40، أي المقاس الذي يجب أن يخزّن المتجر منه أكثر.

والآن يوم متجر ثانٍ:

$$
38,\;39,\;40,\;41,\;42
$$

كل قيمة تظهر مرة واحدة. **لا يوجد منوال** — وهذا نتيجة مفيدة: لا مقاس يهيمن، فخزّن بالتساوي بدلًا من التكديس على مقاس واحد.

## البيانات ثنائية المنوال

أزمنة الانتظار في عيادتين، مدمجة في قائمة واحدة:

$$
5,\;6,\;6,\;7,\;8,\;30,\;31,\;31,\;32,\;33
$$

هناك عنقودان، أحدهما حول 6 دقائق والآخر حول 31. القائمة **ثنائية المنوال** (bimodal). المتوسط نحو 18.9، وهو انتظار لم يعشه أحد فعلًا؛ والوسيط نحو 19.5، وهو غير ممثّل أيضًا. أما المنوالان (6 و31) فهما الملخصان الوحيدان المطابقان لما يحدث.

## الشيء نفسه في بايثون


```python
sizes = [38, 39, 39, 40, 40, 40, 41, 42, 42]

from collections import Counter
counts = Counter(sizes)
print(counts.most_common(1))       # [(40, 3)]

from statistics import multimode
print(multimode([38, 39, 40, 41, 42]))   # [] — no mode at all
print(multimode([1, 1, 2, 2, 3]))        # [1, 2] — two modes
```

`Counter` هو الطريقة الصادقة لرؤية الشكل قبل تلخيصه؛ و`multimode` يعيد كل قيمة متعادلة بدلًا من اختيار واحدة بصمت.

## متى تستخدمه

- **البيانات الفئوية**: الألوان والعلامات التجارية والإدارات وأنواع الأجهزة — كل ما لا ترتيب له ولا حساب.
- **البيانات المنفصلة ذات القيمة المهيمنة**: مقاس الحذاء الأكثر شيوعًا، ورمز الخطأ الأكثر تكرارًا.
- **العثور على مجموعات فرعية** قبل التلخيص: المنوالان إشارة إلى تقسيم البيانات.

## القيود

- **قد يغيب** (لا شيء يتكرر) أو يكون غامضًا (تعادلات كثيرة)، فيصبح عديم الفائدة كملخص واحد في تلك الحالات.
- **يتجاهل كل شيء عدا القمة.** تغيّر صغير في قيمة واحدة قد يخلق تعادلًا أو يزيله.
- **بالنسبة للبيانات المتصلة غير مستقر.** يغيّر التقريب وعرض الحزم الإجابة، لذا نادرًا ما يُستخدم للقياسات.
- **لا يقول شيئًا عن بقية التوزيع.**

## أخطاء شائعة

- **الإبلاغ عن "لا منوال" كخطأ.** إنها نتيجة صالحة وتستحق الذكر.
- **استخدام المنوال للبيانات الرقمية لأنه سهل.** للقياسات، يحمل المتوسط أو الوسيط معلومات أكثر بكثير.
- **تجاهل منوال ثانٍ** عندما يكون أكثر ما يثير الاهتمام في البيانات.
