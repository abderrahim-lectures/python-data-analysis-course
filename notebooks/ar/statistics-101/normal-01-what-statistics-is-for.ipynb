## ماذا يعني

**الإحصاء** (statistics) هو تخصص الإجابة عن الأسئلة باستخدام بيانات ناقصة وغير مؤكدة.

لهذه الجملة ثلاثة أجزاء تستحق الفصل، لأن كل جزء منها موضع خطأ من نوع مختلف.

- **الإجابة عن الأسئلة.** إن لم يكن ثمة سؤال فلا إحصاء — مجرد حساب. "ما متوسط هذه الأرقام؟" حساب. أما "هل نظام التوجيه الجديد أسرع *في العموم*؟" فسؤال إحصائي، لأنك لا ترى إلا بعض التوصيلات دائمًا.
- **استخدام البيانات.** يجب أن تأتي الإجابة من مشاهدات لا من رأي أو سلطة.
- **ناقصة وغير مؤكدة.** لا ترى كل شيء أبدًا. عينة، أو شهر من سنة، أو ثلاثون قطعة من مليون. ولذلك تحمل كل نتيجة تحفظًا عن مقدار خطئها المحتمل.

للإحصاء عائلتان:

| | وصفي | استدلالي |
|---|---|---|
| السؤال | كيف تبدو *هذه* البيانات؟ | ماذا تقول هذه البيانات عن العالم الأوسع؟ |
| الأدوات | المتوسط والوسيط والتشتت والمخططات | الخطأ المعياري وفترات الثقة والاختبارات |
| الخطر | التلخيص السيئ | التعميم الخاطئ |

النصف الأول من هذا المقرر وصفي؛ والنصف الثاني استدلالي. ولا أي منهما أصعب، لكنهما يجيبان عن سؤالين مختلفين، والثاني يستند دائمًا إلى الأول.

## لماذا يضيف شيئًا

دون الإحصاء لا يبقى لك إلا الحكايات. "عميلنا الأخير كان راضيًا" و"معظم عملائنا راضون" ادعاءان مختلفان، والثاني وحده يمكن أن يسند قرارًا. الإحصاء هو الآلة التي تحول كومة مشاهدات إلى ادعاء تستطيع الدفاع عنه، ويخبرك — وهذا مهم بالقدر نفسه — متى تكون الكومة أصغر أو أقل تمثيلًا من أن تدافع عن أي شيء.

## الكلمات الست التي يستخدمها كل ما يأتي

تعلّمها الآن؛ فكل درس لاحق يستخدمها دون تعريف.

- **المجتمع** (population) — المجموعة الكاملة من الأشياء التي تريد الحديث عنها. كل التوصيلات هذا العام. كل البالغين في البلد. كل قطعة ستصنعها المصنع يومًا.
- **العينة** (sample) — المجموعة الفرعية التي رصدتها فعلًا. التوصيلات الخمسة والعشرون التي قست زمنها. الأشخاص الأربعمائة الذين استطلعتهم.
- **المعلمة** (parameter) — رقم يصف **المجتمع**. متوسط زمن التوصيل الحقيقي. مجهولة عادةً، وهي الشيء الذي تحاول تقديره.
- **الإحصاءة** (statistic) — رقم يصف **العينة**. متوسط توصيلاتك الخمس والعشرين. معلومة، لكنها تتأرجح.
- **المتغير** (variable) — خاصية يمكن أن تختلف بين الحالات. زمن التوصيل، وتقييم الرضا، وهل تضررت الحزمة.
- **المشاهدة** (observation) — شيء واحد قيس. توصيل واحد، شخص واحد، قطعة واحدة.

زوج المعلمة والإحصاءة هو الأهم. فالمعلمة ثابتة ومجهولة؛ والإحصاءة معلومة ومتغيرة. وكل طريقة استدلالية في هذا المقرر هي سبيل للانتقال من الثاني إلى الأول.

## الوصفي والاستدلالي في صورة واحدة

تقيس 25 توصيلًا وتحسب متوسطًا مقداره 18 دقيقة. ذلك وصف لبياناتك.

ثم تسأل هل متوسط *كل* التوصيلات قريب من 18 على الأرجح، وتلحق به مدى. ذلك استدلال.

الخطوة الثانية مشروعة فقط إذا سُحبت العينة دون تحيز. فالفترة المثالية حول عينة متحيزة تصف مجتمعًا لن تلتقيه أبدًا. ولهذا يهم تصميم جمع البيانات أكثر من تطور التحليل.

## ما لا يستطيع الإحصاء فعله

- **لا يحوّل سؤالًا سيئًا إلى سؤال جيد.** "هل منتجنا جيد؟" لا إجابة قابلة للقياس له حتى يُعرَّف "جيد" كمتغير.
- **لا يصلح عينة متحيزة.** إعادة معاينة التحيز تعيد إنتاجه بدقة.
- **لا يثبت السببية من بيانات رصدية وحدها**، مهما كان الترابط قويًا أو مخرجات البرمجية فاخرة.
- **لا يختلق معلومة لم تُسجل قط.** المتغير المفقود يبقى مفقودًا؛ لا طريقة تستعيده.

## الشيء نفسه في بايثون

يظهر التمييز في الكود الذي تكتبه. وصف عينة سطر واحد:


```python
deliveries = [8, 12, 10, 15, 45]
mean = sum(deliveries) / len(deliveries)
print(mean)            # 18.0 — إحصائي يخص العينة
```

أما القول بشيء عن المجتمع فيتطلب أكثر: خطأ معياري، وفترة، وافتراضًا عن كيفية سحب العينة. وذلك موضوع الوحدة 4 وما بعدها.

## متى تستخدمه

هذه المفردات لكل حديث عن البيانات. عندما يقول أحدهم "المستخدم المتوسط"، اسأل: متوسط أي عينة، بالنيابة عن أي مجتمع؟

## القيود

- **التعريفات دقيقة والعالم ليس كذلك.** تحديد ما يعد مجتمعًا حكم تقديري، والحكم المختلف يعطي إجابة مختلفة.
- **فصل الوصفي عن الاستدلالي ليس ترتيبًا للجودة.** وصف دقيق لمجموعة بيانات كاملة أنفع من استدلال مهمل.
- **المفردات لا تخبرك كيف تصمم دراسة.** ذلك من شأن المجال والتصميم التجريبي.

## أخطاء شائعة

- **قول "المجتمع" وأنت تقصد "العينة"**، مما يجعل كل عبارة عن عدم اليقين خاطئة بعد ذلك.
- **معاملة عينة مريحة كأنها عشوائية.** من أجاب نادرًا ما يكون من أردت.
- **الإبلاغ عن إحصاءة وتسميتها معلمة.** "المتوسط 18" ادعاء عن عينتك؛ ومتوسط المجتمع مجهول.
- **تخطي السؤال.** "لنحلل البيانات" دون سؤال ينتج مخططات لم يطلبها أحد.
