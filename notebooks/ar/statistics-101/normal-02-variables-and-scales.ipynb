## ماذا يعني

**المتغير** (variable) خاصية يمكن أن تأخذ قيمًا مختلفة عبر الحالات. قبل أن تلخصه، تحتاج إلى معرفة نوعه، لأن النوع يحدد أي حساب له معنى.

الفصل الأول هو **فئوي مقابل رقمي**:

- القيم **الفئوية** تسمّي مجموعة: اللون، والقسم، ونعم/لا، ونوع الجهاز. تستطيع عدّها وإيجاد الأكثر شيوعًا، لكن "متوسط اللون" عديم المعنى.
- القيم **الرقمية** أعداد يكون الحساب عليها ذا معنى: الزمن، والسعر، والعدد، ودرجة الحرارة.

تنقسم المتغيرات الرقمية مرة أخرى حسب **المقياس** (scale)، الذي يسجل مقدار البنية التي تحملها الأعداد.

| المقياس | البنية | أمثلة | الملخصات الصالحة |
|---|---|---|---|
| **اسمي** (nominal) | أسماء فقط، بلا ترتيب | اللون، المدينة، نعم/لا | المنوال، والتعدادات، والنسب |
| **ترتيبي** (ordinal) | أسماء بترتيب ذي معنى | صغير/متوسط/كبير، رضا من 1 إلى 5، مقياس الألم | الوسيط، والمنوال، والمدى الربيعي |
| **فتري** (interval) | مرتب، بفواصل متساوية، بلا صفر حقيقي | درجة الحرارة بالمئوية، السنة، درجة IQ | المتوسط، والوسيط، والانحراف المعياري |
| **نسبي** (ratio) | مرتب، بفواصل متساوية، بصفر حقيقي | الزمن، والسعر، والوزن، والعدد | كل شيء، بما في ذلك النسب |

القاعدة العملية: **المقياس يحدد أي المقاييس مسموح.** متوسط متغير اسمي خطأ. ونسبة قيمتين فتريتين ("اليوم أحر مرتين من الأمس") هراء، لأن 0°مئوية ليس غيابًا لدرجة الحرارة.

## لماذا يضيف شيئًا

تكاد كل قصة "الإحصاءات خاطئة" في الصحافة ترجع إلى متغير استُخدم على المقياس الخطأ. ومتوسط تقييمات النجوم مدافَع عنه فقط إذا قبلت معاملة مقياس ترتيبي كفتري. واقتباس "ارتفعت الحرارة 50%" عديم المعنى دون ذكر الوحدات. وترميز الأقسام بـ1 و2 و3 ثم حساب متوسط الرموز ينتج رقمًا لا يعني شيئًا على الإطلاق.

معرفة المقياس تخبرك أي سؤال يُسمح لك بطرحه أصلًا.

## مثال محسوب

مجموعة بيانات تذاكر دعم:

| العمود | القيم | النوع | المقياس |
|---|---|---|---|
| ticket_id | T-1001, T-1002 | فئوي | اسمي |
| priority | low, medium, high | فئوي | **ترتيبي** |
| response_minutes | 3, 45, 1200 | رقمي | **نسبي** |
| satisfaction | 1, 2, 3, 4, 5 | رقمي | **ترتيبي** |
| agent_id | 12, 47, 3 | يشبه الرقمي | **اسمي** |
| opened_at | 2026-03-01 | تاريخ | فتري |

**قراءة الجدول.**

- `ticket_id` تسمية تصادف أنها تحتوي أرقامًا. وترتيبها لا يعني شيئًا.
- لـ`priority` ترتيب، فالوسيط ذو معنى، لكن الفجوات بين المنخفض والمتوسط لا تساوي الفجوات بين المتوسط والمرتفع، فالمتوسط مجازفة.
- لـ`response_minutes` صفر حقيقي (استجابة فورية) ووحدات متساوية، فكل مقياس وكل نسبة صالحان. و"أبطأ مرتين" عبارة حقيقية.
- `satisfaction` هو الفخ الكلاسيكي: الأعداد من 1 إلى 5 تسميات مرتبة لا قياسات. وكثير من الفرق يحسب متوسطها، وهو اصطلاح لا حقيقة — ويستحق أن يُفعل بوعي.
- `agent_id` هو الفخ الأشد. إنه مخزّن كعدد، فتحسب البرمجية متوسطه بارتياح، والإجابة عديمة المعنى.
- `opened_at` فتري: الفرق بين تاريخين ذو معنى، لكن "2026 ضعف 1013" ليس كذلك.

## كيف يغيّر المقياس التحليل

السؤال نفسه — "هل الرضا أعلى لدى الفريق أ؟" — له إجابات مختلفة حسب المقياس:

- على متغير **اسمي**، تقارن النسب ("نسبة أكبر من عملاء أ اختارت 'راضٍ جدًا'")؛
- على متغير **ترتيبي**، تقارن الوسيطات أو التوزيعات، أو تستخدم اختبار رتب؛
- على متغير **فتري أو نسبي**، يكون فرق المتوسط واختبار t مناسبين.

اختيار المقياس الخطأ لا يغيّر الرقم فحسب؛ بل قد يقلب الاستنتاج. ولهذا يأتي درس المفردات قبل الحساب.

## الشيء نفسه في بايثون


```python
import pandas as pd

df = pd.DataFrame({
    'priority': pd.Categorical(['low', 'high', 'medium'], categories=['low', 'medium', 'high'], ordered=True),
    'response': [3, 45, 1200],
    'agent_id': [12, 47, 3],
})

print(df['priority'].cat.ordered)      # True — an ordered categorical
print(df['response'].mean())           # legal: ratio scale
print(df['agent_id'].mean())           # runs, but meaningless — nominal
```

ستحسب pandas متوسطًا لـ`agent_id` دون تحذير. لا يحميك نظام الأنواع هنا؛ يحميك المقياس، وبشرط أن تفحصه.

## القيود

- **الحدود حكم تقديري أحيانًا.** هل تقييم الرضا من 1 إلى 10 ترتيبي أم فتري؟ يصح اعتباره ترتيبيًا، ويُعامل عادةً كفتري. أعلن اختيارك.
- **قد يتغير المقياس مع السؤال.** العمر بالسنوات نسبي؛ وفئة العمر ترتيبية.
- **البيانات الحقيقية تخلط الأنواع في عمود واحد غالبًا** — عمود رقمي فيه `"n/a"` رقمي ومفقود معًا.
- **التصنيف لا يخبرك بجودة البيانات**، بل بما هو حساب مشروع.

## أخطاء شائعة

- **حساب متوسط الرموز.** رموز البريد وأرقام الأقسام ومعرّفات الموظفين اسمية.
- **معاملة مقياس ترتيبي كفتري دون إعلان ذلك.**
- **حساب النسب على بيانات فترية.** "ضعف IQ" و"ضعف درجة الحرارة" بلا معنى.
- **تسمية متغير رقميًا لأنه مخزّن كرقم.** افحص معنى القيم لا نوع تخزينها.
