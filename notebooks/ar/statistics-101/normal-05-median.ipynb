## ماذا يعني

**الوسيط** (median) هو القيمة الوسطى بعد ترتيب البيانات. نصف القيم يساويها أو يقل عنها، ونصفها يساويها أو يزيد عليها. الموضع هو كل ما يهم — حجم القيم القريبة من الوسط لا يهم، وكذلك حجم القيم عند الأطراف.

لـ$n$ قيمة مرتبة تصاعديًا $x_{(1)} \le x_{(2)} \le \cdots \le x_{(n)}$:

$$
\text{median} =
\begin{cases}
x_{\left(\frac{n+1}{2}\right)} & n \text{ odd} \\[4pt]
\dfrac{x_{\left(\frac{n}{2}\right)} + x_{\left(\frac{n}{2}+1\right)}}{2} & n \text{ even}
\end{cases}
$$

بالكلمات: مع عدد فردي يوجد وسط واحد بالضبط؛ ومع عدد زوجي توجد قيمتان وسطيتان، وتأخذ متوسطهما.

## لماذا يضيف شيئًا

يمكن لقيمة واحدة أن تسحب المتوسط. أما الوسيط فلا تستطيع *أي* قيمة منفردة سحبه، مهما كانت متطرفة، لأن نقل قيمة إلى 1,000,000 لا يغيّر أي عنصر يجلس في الوسط. يجيب الوسيط عن سؤال مختلف: لا "ما المجموع الموزّع؟" بل "كيف تبدو الحالة النموذجية؟"

هذا يجعل الوسيط الملخص الصادق لكل ما فيه التواء: الدخول، وأسعار المنازل، وأزمنة الانتظار في المستشفيات، وأحجام الملفات، وأزمنة الاستجابة. في كل واحد من هذه توجد حالات قليلة كبيرة جدًا ستهيمن على المتوسط.

## مثال محسوب يدويًا، عدد فردي

أزمنة التوصيل مرة أخرى، مرتبة:

$$
8,\; 10,\; 12,\; 15,\; 45
$$

هناك $n = 5$ قيم، فالوسط هو الموضع $(5+1)/2 = 3$:

$$
x_{(3)} = 12
$$

الوسيط 12 دقيقة. قارن ذلك بالمتوسط، 18. الوسيط هو ما يسميه العميل توصيلًا عاديًا؛ والمتوسط يحمل ثقل التوصيل الذي استغرق 45 دقيقة.

## مثال محسوب يدويًا، عدد زوجي

أسقط أبطأ توصيل:

$$
8,\; 10,\; 12,\; 15
$$

الآن $n = 4$، والموضعان الوسطيان هما $4/2 = 2$ و$3$:

$$
\text{median} = \frac{x_{(2)} + x_{(3)}}{2} = \frac{10 + 12}{2} = 11.
$$

لاحظ ما حدث: مع عدد زوجي قد يكون الوسيط قيمة لا توجد في البيانات أصلًا (11 ليس واحدًا من الأزمنة الأربعة). ومع ذلك يبقى الوسط الصحيح.

## الشيء نفسه في بايثون


```python
times = [8, 12, 10, 15, 45]
ordered = sorted(times)
n = len(ordered)

if n % 2 == 1:
    median = ordered[n // 2]
else:
    median = (ordered[n // 2 - 1] + ordered[n // 2]) / 2

print(median)      # 12
```

تفعل المكتبة القياسية الشيء نفسه، بما في ذلك متوسط الحالة الزوجية:


```python
from statistics import median

print(median([8, 12, 10, 15, 45]))   # 12
print(median([8, 10, 12, 15]))       # 11.0
```

## قراءة الفجوة بينهما

عادة مفيدة: احسب الاثنين، ثم انظر إلى الفرق.

| البيانات | المتوسط | الوسيط | ما تقوله الفجوة |
|---|---|---|---|
| 8، 10، 12، 15، 45 | 18 | 12 | ذيل أيمن طويل: قيم كبيرة قليلة |
| 10، 11، 12، 12، 13 | 11.6 | 12 | متماثلة تقريبًا: المتوسط والوسيط يتفقان |
| −5، 10، 11، 12، 12 | 8 | 11 | ذيل أيسر طويل: قيم صغيرة قليلة |

عندما يكون المتوسط أعلى بكثير من الوسيط، تسحبه قيم كبيرة قليلة إلى أعلى. وعندما يكون أدنى بكثير، تسحبه قيم صغيرة قليلة إلى أسفل. وعندما يتفقان تقريبًا، تكون البيانات قريبة من التماثل وأي المقياسين يفي بالغرض.

## رؤيته هندسيًا

الوسيط هو النقطة التي تقسم العدد إلى نصفين، لا النقطة التي توازن الأثقال. اسحب النقطة اليمنى القصوى بعيدًا في [مستكشف المتوسط والوسيط من GeoGebra](https://www.geogebra.org/m/zsbf4jqm) وراقب: يقفز المتوسط، ويبقى الوسيط حيث ينقسم العدد 50/50.

## متى تستخدمه

- تكون البيانات **ملتوية** أو تحتوي قيمًا شاذة لا يمكنك تبرير إزالتها.
- تريد وصف حالة **نموذجية** لا مجموع.
- يكون المتغير **ترتيبيًا**: تقييمات مثل "ضعيف، مقبول، جيد" لها ترتيب ذو معنى لكن لا حساب ذا معنى، فالوسيط صالح حيث لا يكون المتوسط كذلك.

## القيود

- **يتجاهل معظم البيانات.** مجموعتا بيانات مختلفتان جدًا قد تتشاركان الوسيط.
- **لا يتجمع.** وسيط مجموعة مدمجة ليس دالة بسيطة للوسيطين، ولهذا تهيمن المتوسطات على الصيغ اللاحقة.
- **يحتاج قيمة واحدة على الأقل.** مع عدد زوجي قد يبلغ عن رقم لا تملكه أي حالة فعلًا.
- **أقل استقرارًا في العينات الصغيرة جدًا.** مع ثلاث قيم، يمكن لعنصر واحد متغير أن يحركه كثيرًا.

## أخطاء شائعة

- **نسيان الترتيب.** وسيط ترتيب القائمة الخام بلا معنى؛ الترتيب خطوة أولى، في كل مرة.
- **التنصيف على نحو خاطئ في الحالة الزوجية.** احسب متوسط العنصرين الوسطيين، لا العنصر الأوسط والأخير.
- **افتراض أن "الوسيط" يعني "الأكثر شيوعًا".** ذاك هو المنوال، مقياس مختلف.
