## ماذا يعني

**التوزيع الطبيعي** (normal distribution) عائلة من منحنيات جرسية متماثلة، يحدد كل واحد منها عددان: مركزه $\mu$ وتشتته $\sigma$. لكثافته شكل محدد،

$$
f(x) = \frac{1}{\sigma\sqrt{2\pi}}\, e^{-\frac{(x-\mu)^2}{2\sigma^2}},
$$

لكنك نادرًا ما تحتاج إلى تقييمه يدويًا. المهم هو نتيجة الشكل على المسافات.

لأن المنحنى متماثل وتشتته يقاس بـ$\sigma$، فإن للمسافات عن المتوسط احتمالات ثابتة:

- نحو **68%** من القيم تقع داخل $1\sigma$ من المتوسط؛
- ونحو **95%** داخل $2\sigma$؛
- ونحو **99.7%** داخل $3\sigma$.

هذه هي قاعدة 68–95–99.7، وهي سبب استحقاق الانحراف المعياري للإبلاغ عنه.

## لماذا يضيف شيئًا

المنحنى الطبيعي هو النموذج المرجعي الذي يحوّل قياسًا خامًا إلى عبارة عن الندرة. وزن طفل 3.1 كغ لا يعني شيئًا وحده؛ والوزن نفسه عند 1.9 انحراف معياري تحت المتوسط وضع سريري مختلف تمامًا.

وهو أيضًا التوزيع وراء الخطأ المعياري ومعظم صيغ الفترات في الدروس اللاحقة، لأن متوسطات أشياء مستقلة كثيرة تميل إلى شكل طبيعي مهما كان شكل البيانات الأصلية. تلك هي نظرية الحد المركزي، ولهذا يظهر هذا المنحنى حتى حين لا تكون البيانات نفسها جرسية بأي شكل.

## مثال محسوب يدويًا: درجة z

لأطوال البالغين في مجتمع ما متوسط $\mu = 170$ سم وانحراف معياري $\sigma = 10$ سم. طول شخص 185 سم. ما مدى غرابة ذلك؟

**الخطوة 1 — عيّر.**

$$
z = \frac{x - \mu}{\sigma} = \frac{185 - 170}{10} = 1.5
$$

**الخطوة 2 — اقرأ الموضع.** الطول يقع على بعد 1.5 انحراف معياري فوق المتوسط. بموجب قاعدة 68–95–99.7، يقع 68% من الناس بين $160$ و$180$ سم، إذن 32% خارج ذلك؛ ونصفهم فوق 180، أي نحو 16%. صاحبنا أبعد من 180 لكن أقل بعدًا من 190 (انحرافان معياريان)، إذن نحو 7% من الناس أطول منه.

**الخطوة 3 — قلها بالكلمات.** "نحو 7 من كل 100 بالغ أطول من هذا." تلك هي الجملة التي يعطيك إياها جدول أو دالة مكتبة بدقة، وتقاربها القاعدة دونها.

## الشيء نفسه في بايثون


```python
import statistics as st

heights = [158, 162, 165, 167, 170, 171, 174, 176, 179, 188]
mu = st.mean(heights)
sigma = st.pstdev(heights)      # صيغة المجتمع، كما في النموذج أعلاه

z = (185 - mu) / sigma
print(round(mu, 1), round(sigma, 2), round(z, 2))
```

للحصول على نسب دقيقة، تعطي دالة التوزيع التراكمي الطبيعي القياسي المساحة تحت درجة z:


```python
from statistics import NormalDist

normal = NormalDist(mu=170, sigma=10)
print(normal.cdf(185))              # ≈ 0.9332، النسبة دون 185
print(1 - normal.cdf(185))          # ≈ 0.0668، النسبة فوق 185
```

`NormalDist` جزء من مكتبة بايثون القياسية، فلا يحتاج هذا إلى أي تثبيت.

## رؤيته هندسيًا

المساحة تحت المنحنى هي الاحتمال، وقاعدة 68–95–99.7 مجرد عبارة عن المساحات بين خطوط عمودية. في [أداة التوزيع الطبيعي من GeoGebra](https://www.geogebra.org/m/yjz6bqtn) يمكنك ضبط المتوسط والانحراف المعياري وسحب الحد لمراقبة تغير المساحة المظللة — بما في ذلك اللحظة التي تنتقل فيها من انحراف معياري إلى اثنين فتنكمش المساحة من الثلث إلى العشرين.

## متى تستخدمه

- تكون البيانات **متماثلة وحيدة القمة**، وتريد نسب أذيال تقريبية.
- تعمل مع **متوسطات مشاهدات كثيرة**، حيث تنطبق نظرية الحد المركزي حتى لو كانت البيانات الخام ملتوية.
- تحتاج إلى **مقياس معياري** لمقارنة قياسات بوحدات مختلفة، وهذا بالضبط ما تقدمه درجات z.

## القيود

- **معظم البيانات الحقيقية ليست طبيعية.** الدخول والأسعار وأزمنة الانتظار وأعداد السكان والتعدادات ملتوية؛ وفرض منحنى جرسي عليها ينتج هراءً واثقًا.
- **غير محدود.** يسمح النموذج بأطوال سالبة وأزمنة انتظار سالبة، وهذا لا يحدث.
- **الأذيال أنحف من الواقع.** تنتج البيانات الحقيقية أحداثًا متطرفة أكثر بكثير مما يتنبأ به المنحنى الطبيعي.
- **لا يقول شيئًا عن الميكانيزم.** الشكل الجرسي وصف لا تفسير.

## أخطاء شائعة

- **استخدام الانحراف المعياري للعينة حيث ينبغي استخدام انحراف المجتمع** (أو العكس عند بناء درجات z)، مما يزيح كل احتمال قليلًا.
- **افتراض الطبيعية لأن إحصاءات الملخص تبدو معقولة.** ارسم البيانات أولًا؛ وافحص الالتواء وتعدد القمم.
- **اقتباس درجات z لبيانات شديدة الالتواء** وتفسيرها كنسب.
