## ماذا يعني

يقسم الوسيط البيانات المرتبة إلى نصفين. قسّم كل نصف مرة أخرى، فتحصل على **الربيعيات** (quartiles): ثلاث نقاط قطع تقسم البيانات إلى أربع مجموعات متساوية الحجم تقريبًا.

- $Q_1$، الربيع الأول: 25% من القيم تساويه أو تقل عنه.
- $Q_2$، الربيع الثاني: الوسيط، 50% تحته.
- $Q_3$، الربيع الثالث: 75% تحته.

**المدى الربيعي** (interquartile range) هو المسافة بين الطرفين:

$$
\text{IQR} = Q_3 - Q_1 .
$$

إنه عرض النصف الأوسط من البيانات — المدى الذي يحتوي الحالة النموذجية ويتجاهل الذيلين كليهما.

## لماذا يضيف شيئًا

يربّع الانحراف المعياري مسافاته، فقيمة متطرفة واحدة قد تهيمن عليه. أما المدى الربيعي فيستخدم المواضع، تمامًا كالوسيط، فلا تستطيع أي قيمة منفردة — مهما كانت متطرفة — تحريكه. الجأ إلى المدى الربيعي عندما تكون أذيال البيانات طويلة، أو حين تشك في أن بعض العناصر خاطئة لكنك لم تتأكد.

يمنح المدى الربيعي أيضًا طريقة مبدئية للقول "هذه القيمة بعيدة على نحو مريب": **قاعدة 1.5 × المدى الربيعي**. يُعلَّم أي شيء يقع دون $Q_1 - 1.5\,\text{IQR}$ أو فوق $Q_3 + 1.5\,\text{IQR}$. إنه اصطلاح لا قانون، لكنه الاصطلاح الذي ترسمه مخططات الصندوق.

## مثال محسوب يدويًا

أحد عشر زمن انتظار مرتبة بالدقائق:

$$
2,\;4,\;4,\;5,\;7,\;9,\;11,\;12,\;14,\;18,\;40
$$

**الخطوة 1 — الوسيط.** مع $n=11$ يكون الموضع الأوسط 6، إذن $Q_2 = 9$.

**الخطوة 2 — القسمة إلى نصفين.** تحت الوسيط: $2,4,4,5,7$. وفوقه: $11,12,14,18,40$.

**الخطوة 3 — الربيعيات هي وسيطا هذين النصفين.** كل نصف فيه خمس قيم، إذن $Q_1 = 4$ و$Q_3 = 14$.

**الخطوة 4 — المدى الربيعي.**

$$
\text{IQR} = 14 - 4 = 10 \text{ minutes}
$$

**الخطوة 5 — سياج القيم الشاذة.** السياج الأدنى $4 - 15 = -11$؛ والسياج الأعلى $14 + 15 = 29$. القيمة 40 تقع فوق 29، فتُعلَّم. الانحراف المعياري لهذه البيانات نحو 10.6، وقد نفخته القيمة 40 نفسها؛ أما المدى الربيعي فلم يتحرك.

لاحظ أن الأسوار ليست تعليمات "احذف هذه القيمة". إنها تعلّم قيمة تستحق الفحص: خطأ مطبعي، أو وحدة مختلفة، أو حالة غريبة فعلًا.

## ملخص الخمسة أعداد ومخطط الصندوق

تصف خمسة أعداد الشكل دون أي افتراض عنه:

$$
\text{min},\; Q_1,\; Q_2,\; Q_3,\; \text{max}
= 2,\; 4,\; 9,\; 14,\; 40 .
$$

يرسم مخطط الصندوق هذه بالضبط: صندوق من $Q_1$ إلى $Q_3$ بخط عند الوسيط، وشعيرات إلى القيم المتطرفة داخل الأسوار، ونقاط لما يقع بعدها. عرض الصندوق *هو* المدى الربيعي، فيجعل المخطط التشتت مرئيًا بنظرة.

يمكنك بناء واحد وسحب القيم في [أداة مخطط الصندوق من GeoGebra](https://www.geogebra.org/m/b5cct2aq)؛ ومراقبة الصندوق يثبت بينما تمتد شعيرة هي أوضح طريقة لرؤية ما يتحمله المدى الربيعي.

## الشيء نفسه في بايثون


```python
times = [2, 4, 4, 5, 7, 9, 11, 12, 14, 18, 40]

import statistics as st
q1 = st.quantiles(times, n=4)      # [4, 9, 14]
iqr = q1[2] - q1[0]
print(q1, iqr)                      # [4, 9, 14] 10

low, high = q1[0] - 1.5 * iqr, q1[2] + 1.5 * iqr
print(low, high)                    # -11.0 29.0
print([t for t in times if t < low or t > high])   # [40]
```

مع pandas تأتي الكميات نفسها من `describe()`، التي تطبع ملخص الخمسة أعداد زائد المتوسط والانحراف المعياري في استدعاء واحد:


```python
import pandas as pd

s = pd.Series(times)
print(s.describe())
```

## متى تستخدمه

- **البيانات الملتوية**: الدخول والأسعار وأزمنة الاستجابة، وكل ما له ذيل طويل.
- **البيانات المريبة**: تجد قاعدة السياج عناصر تستحق الفحص قبل أن تشوّه متوسطًا.
- **مقارنة المجموعات** بأحجام عينات مختلفة، حيث يكون مخطط صندوق لكل مجموعة أكثر إفادة من متوسطين.

## القيود

- **يلقي الأذيال بعيدًا.** مجموعتا بيانات قد تتشاركان ملخص الخمسة أعداد وتختلفان في كل شيء آخر.
- **تعريفات الربيعيات تختلف.** تحسب البرمجيات المختلفة $Q_1$ و$Q_3$ على نحو مختلف قليلًا للعينات الصغيرة، فإجابة يدوية قد تكون خطوة بعيدة عن إجابة حزمة. المدى الربيعي مستقر بما يكفي للاستخدام العملي.
- **لا حساب.** كالوسيط، لا يتجمع عبر المجموعات.
- **العينات الصغيرة تعطي ربيعيات خشنة.** مع أقل من ثماني قيم تقريبًا، لا تكون الربيعيات أكثر إفادة بكثير من الأدنى والأعلى.

## أخطاء شائعة

- **عدم الترتيب أولًا.** الربيعيات موضعية كالوسيط؛ والبيانات غير المرتبة تعطي هراءً.
- **قراءة شعيرات مخطط الصندوق كأدنى وأعلى قيمة.** تتوقف الشعيرات عند السياج؛ والنقاط بعدها تُرسم منفصلة.
- **معاملة علامة الشذوذ كخطأ.** إنها دعوة للنظر لا حكم.
- **مقارنة المديات الربيعية عبر وحدات مختلفة.** الدقائق والدولارات غير قابلتين للمقارنة؛ استخدم معامل التغير إن اضطررت لمقارنة التشتت.
