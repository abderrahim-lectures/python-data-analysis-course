## ماذا يعني

**مفارقة سيمبسون** (Simpson's paradox) هي انقلاب اتجاه يظهر عند دمج المجموعات. قد يبدو علاج أفضل من ضابط في كل مجموعة فرعية، وأسوأ عند دمج المجموعات الفرعية. لا شيء محسوب خطأً؛ فالرقم المدمج حقيقي، ويجيب عن سؤال مختلف عن أسئلة الأرقام المجمّعة.

تحدث المفارقة عندما تختلف المجموعات في الحجم *و*في مستوى خط أساسها. فالمجموعة الأكبر تسحب المتوسط المدمج نحو ركنها، وإذا صادف أنها تتلقى أحد العلاجين أكثر، فالعلاج يرث عيب المجموعة.

## لماذا يضيف شيئًا

إنه أوضح مثال على ملخص صحيح حسابيًا وخاطئ جوهريًا. كل محلل يجمّع؛ والمفارقة هي سبب النظر إلى المجموعات قبل الثقة في رقم مدمج. والتعرف عليها يحول خلافًا محيّرًا بين تقريرين إلى تفسير بنيوي.

## مثال محسوب يدويًا

مستشفيان، ومجموعتا خطورة، ومعدلات شفاء:

| المستشفى | الحالات الخفيفة | الشفاء | الحالات الشديدة | الشفاء |
|---|---|---|---|---|
| أ | 90 | 81 (90%) | 10 | 6 (60%) |
| ب | 10 | 8 (80%) | 90 | 45 (50%) |

**الخطوة 1 — قارن داخل كل درجة خطورة.** يشفي المستشفى أ 90% من الحالات الخفيفة مقابل 80% لـب، و60% من الحالات الشديدة مقابل 50% لـب. يفوز أ في المجموعتين.

**الخطوة 2 — ادمج المجاميع.** المستشفى أ: $81 + 6 = 87$ حالة شفاء من 100، إذن 87%. والمستشفى ب: $8 + 45 = 53$ من 100، إذن 53%.

**الخطوة 3 — تحقق من الحساب.** يفوز أ في المجمل أيضًا — إذن هذا الجدول *ليس* المفارقة، بل الحالة العادية. لرؤية الانقلاب، أعطِ ب المجموعة الخفيفة الأكبر:

| المستشفى | الحالات الخفيفة | الشفاء | الحالات الشديدة | الشفاء |
|---|---|---|---|---|
| أ | 20 | 18 (90%) | 80 | 48 (60%) |
| ب | 90 | 72 (80%) | 10 | 5 (50%) |

الآن يشفي أ 90% من الخفيفة مقابل 80% لـب، و60% من الشديدة مقابل 50% لـب — لا يزال أ يفوز في المجموعتين. والمدمج: أ لديه $18+48 = 66$ من 100، أي 66%؛ وب لديه $72+5 = 77$ من 100، أي 77%. **يفوز ب في المجمل بينما يخسر في كل مجموعة فرعية.**

**الخطوة 4 — قل لماذا.** يعالج أ معظم الحالات الشديدة، وهي ذات معدل شفاء منخفض بغض النظر عن المستشفى. ويعالج ب معظم الحالات الخفيفة، وهي ذات معدل مرتفع. فالرقم المدمج يقيس *من يعالج كل مستشفى* بقدر ما يقيس *مدى جودة علاجه لهم*.

## رؤيته هندسيًا

ارسم كل مجموعة نقطةً حيث $x$ = حجم المجموعة و$y$ = معدل الشفاء. المتوسط المدمج متوسط مرجّح، وموضعه يعتمد على مكان الأوزان. في [مستكشف مفارقة سيمبسون من GeoGebra](https://www.geogebra.org/m/dqzxhqzy) يمكنك سحب أحجام المجموعات ومراقبة النقطة المدمجة تعبر الخطوط الفردية.

## الشيء نفسه في بايثون


```python
a = {'mild': (20, 18), 'severe': (80, 48)}
b = {'mild': (90, 72), 'severe': (10, 5)}

for name, g in [('A', a), ('B', b)]:
    for sev, (n, rec) in g.items():
        print(name, sev, f"{rec/n:.0%}")
    total_n = sum(n for n, _ in g.values())
    total_r = sum(r for _, r in g.values())
    print(name, 'pooled', f"{total_r/total_n:.0%}")
```

النسب المجمّعة تفضل أ في الصفين والنسبة المدمجة تفضل ب. قراءة ذلك الناتج هي التمرين كله؛ والكود لا يضيف شيئًا عدا السرعة.

## متى تبحث عنه

- أي مقارنة تكون فيها المجموعات **غير متساوية في الحجم** وتختلف في الخطورة الأساسية.
- الإبلاغ عن **معدل مدمج** من بيانات وصلت في دفعات أو مواقع أو أسابيع أو أفواج.
- نتيجة تناقض توقع خبير في المجال — المفارقة من أول الأشياء التي يجب استبعادها.

## القيود

- **ليست خطأً يحتاج إلى "تصحيح".** المتوسط المدمج رقم مشروع للمجتمع المدمج؛ والسؤال هو أي مجتمع تقصد.
- **قد يكون التحليل المجمّع هو الخاطئ** إذا شرطت على متغير ناتج عن العلاج لا خاصية سابقة.
- **يحتاج متغير تجميع واضح.** إذا لم تعرف المجموعات، فلا يمكنك التعديل لها.

## أخطاء شائعة

- **الإبلاغ عن الرقم المدمج دون المجموعات**، مما يخفي الانقلاب تمامًا.
- **التعديل لمتغير خاطئ**، مما قد يخلق مفارقة بدلًا من إزالتها.
- **افتراض أن العينات الصغيرة محصّنة.** المفارقة حسابية؛ والحجم لا يمنعها.
