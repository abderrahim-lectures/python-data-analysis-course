## ماذا يعني

يكون متغيران **مرتبطين** (correlated) عندما يميلان إلى التحرك معًا. يضغط معامل الارتباط ذلك الميل في رقم واحد بين $-1$ و$+1$.

**ارتباط بيرسون** (Pearson correlation) يقيس مدى قرب النقاط من *خط مستقيم*:

$$
r = \frac{1}{n-1}\sum_{i=1}^{n}
\left(\frac{x_i - \bar{x}}{s_x}\right)
\left(\frac{y_i - \bar{y}}{s_y}\right)
$$

تتحول كل قيمة إلى درجة معيارية — أي عدد الانحرافات المعيارية التي تفصلها عن متوسطها — والصيغة هي متوسط حاصل ضرب الدرجتين المعياريتين. عندما تكون $x$ فوق متوسطها في الوقت نفسه الذي تكون فيه $y$ فوق متوسطها، يكون ذلك الحاصل موجبًا؛ وعندما يختلفان، يكون سالبًا. ومتوسط تلك الحواصل هو $r$.

**ارتباط سبيرمان** (Spearman correlation) هو الصيغة نفسها محسوبة على *الرتب* بدلًا من القيم. هذا يجعله مقاومًا للقيم الشاذة وقادرًا على التقاط أي علاقة تستمر في اتجاه واحد، لا الخطية فقط.

## لماذا يضيف شيئًا

وصف متغير واحد نصف المهمة. كل ادعاء مثير تقريبًا علاقي: علاج مرتبط بنتائج أفضل، وسعر مرتبط بالطلب، ودرجة حرارة مرتبطة بالمبيعات. الارتباط هو أرخص إجابة كمية، ومعرفة حدوده تحميك من أشيع استدلال خاطئ في العمل على البيانات — قراءة سبب في مصادفة.

وهو يغذي أيضًا كل ما يأتي لاحقًا تقريبًا: الانحدار، واختيار الميزات، وفحوص الموثوقية التي تسبق تركيب أي نموذج.

## مثال محسوب يدويًا

خمسة طلاب، ساعات المذاكرة $x$ والدرجة $y$:

$$
x = 1,2,3,4,5 \qquad y = 50,55,65,70,80
$$

**الخطوة 1 — المتوسطان.** $\bar{x} = 3$، $\bar{y} = 64$.

**الخطوة 2 — الانحرافات عن كل متوسط.**

| $x$ | $y$ | $x - \bar{x}$ | $y - \bar{y}$ |
|---|---|---|---|
| 1 | 50 | −2 | −14 |
| 2 | 55 | −1 | −9 |
| 3 | 65 | 0 | +1 |
| 4 | 70 | +1 | +6 |
| 5 | 80 | +2 | +16 |

**الخطوة 3 — الانحرافان المعياريان.** $\sum (x-\bar{x})^2 = 10$، إذن $s_x = \sqrt{10/4} \approx 1.581$. و$\sum (y-\bar{y})^2 = 196 + 81 + 1 + 36 + 256 = 570$، إذن $s_y = \sqrt{570/4} \approx 11.94$.

**الخطوة 4 — عيّر واضرب، صفًا بصف.** للصف الأول، $(-2/1.581)(-14/11.94) \approx 1.265 \times 1.173 \approx 1.484$. نفّذ ذلك للصفوف الخمسة واجمع:

$$
\sum \left(\frac{x-\bar{x}}{s_x}\right)\left(\frac{y-\bar{y}}{s_y}\right) \approx 4.949
$$

**الخطوة 5 — اقسم على $n-1$.**

$$
r = \frac{4.949}{4} \approx 0.987
$$

قريب من 1. ساعات المذاكرة والدرجة ترتفعان معًا بشكل شبه تام في هذه العينة.

**الخطوة 6 — ربّع.** $r^2 \approx 0.975$: 97.5% من تباين الدرجات يسير مع الاتجاه الخطي في ساعات المذاكرة. لاحظ أن $r^2$ هو الرقم الذي تذكره عندما تريد القول "كم منه مفسَّر"؛ أما $r$ نفسه فيبالغ في ذلك.

## قراءة الإشارة والحجم

| $r$ | القراءة |
|---|---|
| $+1$ | زيادة خطية تامة |
| $+0.7$ | اتجاه صاعد قوي، مع تشتت ظاهر |
| $+0.2$ | اتجاه صاعد ضعيف، يكاد يضيع في الضجيج |
| $0$ | لا اتجاه *خطي* (قد يكون ثمة منحنى) |
| $-0.6$ | اتجاه هابط معتدل |

تقول الإشارة الاتجاه، ويقول الحجم مدى إحكام التلاؤم الخطي، ولا شيء أكثر. $r = 0$ ليس "لا علاقة"؛ إنه "لا علاقة خطية" — علاقة على شكل U مقلوب تام لها $r \approx 0$.

## عندما يضللك بيرسون: استخدم سبيرمان

اعتبر هذه النقاط الخمس: $x = 1,2,3,4,5$ و$y = 1,4,9,16,25$. العلاقة تامة وصاعدة، لكنها منحنية. $r$ لبيرسون نحو $0.98$ — مرتفع لكن ليس 1، لأن المنحنى غير مستقيم. ارتباط سبيرمان الرتبي على البيانات نفسها هو 1 بالضبط: كل رتبة في $x$ تطابق رتبة في $y$.

القاعدة العملية:

- استخدم **بيرسون** عندما يكون المتغيران رقميين ومتصلين تقريبًا، ويبدو مخطط الانتشار خطًا أو بيضاويًا؛
- استخدم **سبيرمان** عندما تكون العلاقة رتيبة لكن منحنية، أو حين توجد قيم شاذة، أو عندما يكون أحد المتغيرين ترتيبيًا (تقييم رضا من 1 إلى 5، أو مركز في مسابقة).

## الشيء نفسه في بايثون


```python
x = [1, 2, 3, 4, 5]
y = [50, 55, 65, 70, 80]

import statistics as st
mx, my = st.mean(x), st.mean(y)
sx, sy = st.stdev(x), st.stdev(y)

zx = [(v - mx) / sx for v in x]
zy = [(v - my) / sy for v in y]
r = sum(a * b for a, b in zip(zx, zy)) / (len(x) - 1)
print(round(r, 3))        # 0.987
print(round(r ** 2, 3))   # 0.975
```

تفعل NumPy وpandas الشيء نفسه في استدعاء واحد، و`corr(method='spearman')` يبدّل إلى النسخة الرتبية:


```python
import numpy as np
print(np.corrcoef(x, y)[0, 1])

import pandas as pd
df = pd.DataFrame({'hours': x, 'score': y})
print(df.corr(method='pearson'))
print(df.corr(method='spearman'))
```

## رؤيته هندسيًا

على مخطط الانتشار، يصف $r$ لبيرسون مدى ضيق السحابة حول خط مستقيم. أدِر السحابة فيتغير $r$؛ واثنِها إلى U فيهبط $r$ إلى الصفر بينما تقوى العلاقة. يتيح لك [مستكشف الارتباط من GeoGebra](https://www.geogebra.org/m/gtxfmqsf) سحب النقاط ومراقبة رد $r$، بما في ذلك اللحظة التي تسحب فيها نقطة واحدة خطًا كان محكمًا إلى خط ضعيف.

## متى تستخدمه

- **فحص العلاقات** قبل تحديد ما ستنمذجه.
- **التحقق** من أن ميزة ما تحمل معلومات عن هدف.
- **الإبلاغ** عن الترابط بصدق، إلى جانب مخطط انتشار يراه القارئ بنفسه فيفهم ما يلخصه الرقم.

## القيود

- **الارتباط ليس السببية.** ترتبط مبيعات الآيس كريم بحالات الغرق؛ وكلاهما يتبع الطقس.
- **يقيس الخطوط المستقيمة فقط.** العلاقات المنحنية القوية قد تُظهر $r$ قريبًا من الصفر.
- **حساس للقيم الشاذة.** نقطة واحدة بعيدة قد ترفع $r$ من 0.2 إلى 0.8، أو تدمر ارتباطًا حقيقيًا.
- **المديات المقيدة تصغّره.** الارتباط داخل أعلى 10% من القيم قد يخفي علاقة قوية في المجمل.
- **دمج المجموعات قد يختلق ارتباطًا** (مفارقة سيمبسون): عنقودان بلا اتجاه يمكن أن يظهرا اتجاهًا قويًا عند دمجهما.

## أخطاء شائعة

- **قراءة $r^2$ كنسبة مئوية ثم ذكر $r$ كأنه نسبة مئوية كذلك.** $r = 0.5$ يعني 25% من التباين، لا 50%.
- **مقارنة ارتباطات عبر مجتمعات مختلفة** دون التحقق من أن المديات قابلة للمقارنة.
- **تجاهل مخطط الانتشار.** لا يستطيع الرقم إخبارك عن الانحناء أو العناقيد أو القيم الشاذة؛ الصورة تستطيع.
