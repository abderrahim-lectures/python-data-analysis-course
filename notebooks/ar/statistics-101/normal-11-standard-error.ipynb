## ماذا يعني

يصف **الانحراف المعياري** مدى تشتت *البيانات*. ويصف **الخطأ المعياري** (standard error) مدى تشتت *المتوسط نفسه* لو أعدت الدراسة مرات كثيرة.

$$
\text{SE} = \frac{s}{\sqrt{n}}
$$

حيث $s$ هو الانحراف المعياري للعينة و$n$ حجم العينة.

يجيبان عن سؤالين مختلفين. إذا كان لصف من 30 طالبًا انحراف معياري 12 نقطة في اختبار، فالطلاب يتفاوتون بنحو 12 نقطة. وإذا أعدت الصف مع 30 طالبًا جددًا، فسيتفاوت *متوسط* الصف بنحو $12/\sqrt{30} \approx 2.2$ نقطة فقط. الرقم الأول يتعلق بالأشخاص؛ والثاني يتعلق بتقديرك.

## لماذا يضيف شيئًا

كل متوسط تحسبه هو سحبة واحدة من توزيع من المتوسطات الممكنة. الخطأ المعياري هو عرض ذلك التوزيع، فيخبرك بمقدار الثقة التي توليها للرقم الذي أنت على وشك الإبلاغ عنه.

وهو أيضًا مكوّن كل فترة تقريبًا في الإحصاء التطبيقي: فترة ثقة 95% لمتوسط هي تقريبًا $\bar{x} \pm 2\,\text{SE}$، واختبار t هو نسبة فرق إلى خطأ معياري. فهم هذه الكمية الواحدة يزيل الغموض عن معظم الصيغ التي تلتقيها لاحقًا.

## مثال محسوب يدويًا

لعينة من $n = 25$ زمن توصيل متوسط 18 دقيقة وانحراف معياري للعينة 15.3 دقيقة.

**الخطوة 1 — الجذر التربيعي للعدد n.**

$$
\sqrt{25} = 5
$$

**الخطوة 2 — اقسم.**

$$
\text{SE} = \frac{15.3}{5} = 3.06 \text{ minutes}
$$

**الخطوة 3 — اقرأه.** تتفاوت التوصيلات الفردية بنحو 15 دقيقة، لكن *متوسط* 25 منها غير مؤكد إلا بنحو 3 دقائق. ولو أُعيدت الدراسة بمجموعة جديدة من 25 توصيلًا، فسيقع المتوسط الجديد عادةً ضمن 3 دقائق تقريبًا من 18.

## لماذا الجذر التربيعي

الجذر التربيعي هو أهم جزء في الصيغة، لأنه يحدد سرعة شراء البيانات الإضافية للدقة. لتنصيف الخطأ يجب أن **تضاعف العينة أربع مرات**:

| $n$ | الخطأ المعياري عند $s = 15.3$ |
|---|---|
| 25 | 3.06 |
| 100 | 1.53 |
| 400 | 0.77 |
| 2500 | 0.31 |

عشرة أضعاف البيانات تعطي نحو ثلاثة أضعاف الدقة. ولهذا تتوقف الاستطلاعات عند نحو ألف مستجيب: بعد ذلك يكون المنحنى مسطحًا، ويصبح الخطأ المتبقي مهيمنًا عليه بالتحيز لا بالمعاينة.

## الشيء نفسه في بايثون


```python
import statistics as st

times = [8, 10, 12, 15, 45]
n = len(times)
s = st.stdev(times)
se = s / (n ** 0.5)
print(round(s, 2), round(se, 2))    # 15.31 6.85
```

والتجربة التي تجعل المفهوم ملموسًا: اسحب عينات كثيرة من المجتمع نفسه وقس مدى تفاوت متوسطات العينات. يجب أن يطابق ذلك التشتت الخطأ المعياري.


```python
import random, statistics as st

population = [random.gauss(50, 10) for _ in range(10_000)]
means = [st.mean(random.sample(population, 25)) for _ in range(2000)]

print(round(st.pstdev(means), 2))                                  # ≈ 2.0
print(round(st.pstdev(population) / 25 ** 0.5, 2))                 # ≈ 2.0
```

تطابق الرقمين هو الدرس كله، ويستحق أن تشغّله مرة بعينيك.

## رؤيته هندسيًا

توزيع المعاينة منحنى جرسي مرسوم حول تقديرك، والخطأ المعياري هو نصف عرضه. ضيّقه بزيادة $n$ في [عرض توزيع المعاينة من GeoGebra](https://www.geogebra.org/m/tqjfjzpt) وراقب انقباض المنحنى بمعدل الجذر التربيعي لحجم العينة.

## متى تستخدمه

- الإبلاغ عن **مدى دقة** تقدير ما، كرقم أو كفترة ثقة.
- مقارنة فرق بأرضية **ضجيجه**، وهذا ما تفعله اختبارات الدلالة اللاحقة.
- تخطيط دراسة: حل $\text{SE} = s/\sqrt{n}$ لإيجاد $n$ التي تعطي خطأ مقبولًا.

## القيود

- **يقيس خطأ المعاينة فقط.** إذا كانت العينة متحيزة، فالخطأ المعياري الصغير يجعل الإجابة الخاطئة أدق فقط.
- **يفترض مشاهدات مستقلة.** القياسات المتكررة على الشخص نفسه ليست مستقلة، والصيغة البسيطة تبخس الخطأ.
- **العينات الصغيرة تحتاج توزيع t** لا المعامل الطبيعي 2.0.
- **لا يقول شيئًا عن البيانات نفسها.** خطأ معياري مقداره 1 لا يعني أن القيم الفردية قريبة من المتوسط.

## أخطاء شائعة

- **الخلط بينه وبين الانحراف المعياري.** أشيع خلط في الإحصاء كله: "أشرطة الخطأ صغيرة" غالبًا يصف الخطأ المعياري لمتوسط، لا تشتت الحالات.
- **القسمة على n بدلًا من √n.** يهبط الخطأ أبطأ بكثير مما يتوقع الناس.
- **الإبلاغ عن الخطأ المعياري كمدى ± حول كل قيمة.** إنه مدى للمتوسط لا للبيانات.
