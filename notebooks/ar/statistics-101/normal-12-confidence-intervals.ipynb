## ماذا يعني

**فترة الثقة** (confidence interval) هي مدى من القيم المعقولة لوسيط مجتمع، تُبنى من تقدير عينة زائد ناقص هامش:

$$
\bar{x} \pm t_{n-1,\,0.975}\cdot \frac{s}{\sqrt{n}}
$$

في العينات الكبيرة يقترب المعامل $t$ من 2، ومن هنا يأتي التعبير المألوف "± خطأان معياريان". أما تفسيرها فهو الجزء الدقيق:

> إذا أعدت الدراسة مرات كثيرة، فإن نحو 95% من الفترات التي تبنيها بهذه الطريقة ستحتوي الوسيط الحقيقي.

تصف نسبة 95% **الإجراء** لا فترتك بعينها. أي فترة منفردة إما تحتوي الحقيقة أو لا تحتويها، ولا يمكنك معرفة أيّهما. ما تستطيع قوله هو أن الطريقة صحيحة 95% من الوقت.

## لماذا يضيف شيئًا

يتظاهر التقدير النقطي بدقة لا يملكها. "متوسط زمن التوصيل 18 دقيقة" يبدو دقيقًا وهو شبه مؤكد خاطئ في الرقم الثالث. أما "بين 11.7 و24.3 دقيقة، بثقة 95%" فعبارة تستطيع التصرف بناءً عليها: تخبر المدير أنه لا يستطيع بعد أن يعد بخدمة في 15 دقيقة، لأن 15 داخل المدى المعقول.

تجعل الفترات المقارنات صادقة أيضًا. مجموعتان تتداخل فترتاهما تداخلًا كبيرًا ليستا مختلفتين بوضوح، مهما أوحت متوسطاتهما.

## مثال محسوب يدويًا

عينة من $n=25$ توصيلًا: المتوسط 18 دقيقة، والانحراف المعياري 15.3 دقيقة، والخطأ المعياري 3.06 دقيقة.

**الخطوة 1 — اختر مستوى الثقة.** نسبة 95% هي الاصطلاح. ولـ$n=25$ يكون المعامل الصحيح قيمة $t$ بدرجات حرية 24، وهي نحو 2.064.

**الخطوة 2 — الهامش.**

$$
2.064 \times 3.06 \approx 6.3 \text{ minutes}
$$

**الخطوة 3 — الفترة.**

$$
18 \pm 6.3 \;=\; (11.7,\; 24.3) \text{ minutes}
$$

**الخطوة 4 — قل ماذا يعني.** "نقدّر متوسط زمن التوصيل بـ18 دقيقة، بفترة ثقة 95% من 11.7 إلى 24.3." وعد خدمة بأقل من 12 دقيقة عند حد المعقولية تمامًا؛ ووعد بـ30 يقع فوقه بارتياح.

## ما لا تقوله

أربع قراءات خاطئة شائعة، كلها تستحق التعرف عليها في تقارير الآخرين:

- **"ثمة احتمال 95% أن المتوسط الحقيقي بين 11.7 و24.3."** في القراءة التكرارية، المتوسط الحقيقي رقم ثابت؛ والفترة هي غير المؤكدة.
- **"95% من التوصيلات تستغرق بين 11.7 و24.3 دقيقة."** لا — تلك ستكون فترة تنبؤ، وهي أوسع بكثير.
- **"الفترة تحتوي 95% من البيانات."** إنها تحتوي المتوسط، لا البيانات.
- **"تداخل الفترات يعني عدم وجود فرق."** التداخل تلميح تقريبي لا اختبار؛ فقد تتداخل الفترات بينما يكون الفرق واضحًا.

## الشيء نفسه في بايثون

الحساب اليدوي أولًا، ثم نسخة المكتبة:


```python
import statistics as st

times = [8, 10, 12, 15, 45]
n = len(times)
m = st.mean(times)
se = st.stdev(times) / n ** 0.5
margin = 2.064 * se            # قيمة t لـ 24 درجة حرية عند 95%؛ ولـ n=5 استخدم درجاتك
print(round(m - margin, 1), round(m + margin, 1))
```

تحسب `scipy` المعامل لأي حجم عينة ومستوى ثقة:


```python
from scipy import stats

lo, hi = stats.t.interval(0.95, df=n - 1, loc=m, scale=se)
print(round(lo, 1), round(hi, 1))
```

والتجربة وراء التفسير: ابنِ فترات كثيرة من عينات كثيرة وأحصِ كم مرة تحتوي الحقيقة.


```python
import random, statistics as st
from scipy import stats

population = [random.gauss(50, 10) for _ in range(10_000)]
truth = st.mean(population)
hits = 0
for _ in range(500):
    sample = random.sample(population, 25)
    lo, hi = stats.t.interval(0.95, df=24, loc=st.mean(sample), scale=st.stdev(sample) / 5)
    hits += lo <= truth <= hi
print(hits / 500)      # ≈ 0.95
```

ذلك الناتج — نحو 95% — هو المعنى كله للعبارة.

## رؤيته هندسيًا

تخيّل توزيع المعاينة من الدرس السابق: الفترة هي الـ95% المركزية منه، منقولةً إلى مقياس البيانات. في [عرض فترة الثقة من GeoGebra](https://www.geogebra.org/m/sfqjzhcm) يمكنك تغيير مستوى الثقة ورؤية الفترة تتسع كلما طالبت بيقين أكبر، ثم تضييقها مرة أخرى برفع $n$.

## متى تستخدمه

- الإبلاغ عن أي تقدير تهم دقته: نتائج الاستطلاعات، وأزمنة القياس المرجعي، وقياسات الجودة.
- مقارنة مجموعتين حين تريد إظهار حجم فرق معقول.
- تقرير ما إذا كان قرار آمنًا في أسوأ قيمة معقولة، لا عند أفضل تقدير فقط.

## القيود

- **صالحة فقط إذا كانت المعاينة غير متحيزة.** فترة الثقة حول تقدير متحيز عبارة دقيقة عن الشيء الخطأ.
- **تفترض الاستقلال.** القياسات العنقودية أو المتكررة تحتاج صيغة مختلفة.
- **الفترات الضيقة ليست فترات مهمة.** مع $n$ ضخم، قد يكون لفرق تافه فترة محكمة.
- **ليست فترة تنبؤ.** القيم الفردية أكثر تباينًا بكثير من المتوسط.

## أخطاء شائعة

- **استخدام 1.96 للعينات الصغيرة.** تلك قيمة العينات الكبيرة؛ ومع $n=5$ يكون المعامل الصحيح فوق 2.7.
- **الإبلاغ عن الفترة دون بيان ما هي له.** المتوسط أو الفرق أو النسبة — لكل منها فترته.
- **معاملة عدم التداخل كدليل على الفرق.** إنه موحٍ؛ والمقارنة الصحيحة هي الأداة المناسبة.

- إذا أعطت دراستان فترتين مختلفتين جدًا للكمية نفسها، فما الذي قد يفسر ذلك غير الخطأ؟
- لماذا لا تستطيع فترة الثقة أبدًا إخبارك باحتمال أن فترتك هي الصحيحة؟
- ماذا تحتاج لبناء فترة ثقة للوسيط بدلًا من المتوسط؟
