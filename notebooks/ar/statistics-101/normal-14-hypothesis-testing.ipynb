## ماذا يعني

يسأل **اختبار الفرضيات** (hypothesis test) عن مدى مفاجأة البيانات لو لم يكن ثمة أثر. تصوغ فرضيتين:

- $H_0$، **الفرضية الصفرية** (null hypothesis): لا يوجد فرق، والأثر صفر؛
- $H_1$، **الفرضية البديلة**: يوجد فرق، في اتجاه تحدد مسبقًا.

ثم تحسب إحصاءة تقيس بُعد البيانات المرصودة عما تتوقعه $H_0$، وتحولها إلى **قيمة احتمالية** (p-value) — احتمال بيانات بهذا التطرف على الأقل *لو كانت $H_0$ صحيحة* — وتقارنها بعتبة تحددها سلفًا، وهي 0.05 بالاصطلاح.

لمقارنة متوسطين، الإحصاءة نسبة:

$$
t = \frac{\bar{x}_1 - \bar{x}_2}{\text{SE of the difference}} .
$$

البسط هو الأثر الذي رصدته. والمقام هو مقدار تأرجح ذلك الفرق في ظل $H_0$. والنسبة الكبيرة تعني أن الأثر كبير نسبةً إلى الضجيج.

## لماذا يضيف شيئًا

يمنح إجابة منضبطة عن سؤال "هل يمكن أن يكون هذا محض صدفة؟" — السؤال الذي يثيره كل مقارنة. ويجبرك على تحديد الاتجاه والعتبة قبل رؤية البيانات، وهذا ما يحميك من براعة الإنسان في إيجاد قصة في الضجيج.

وهو أيضًا موطن قدر كبير من الأخطاء الإحصائية المنشورة، فمعرفته دفاعية بقدر ما هي بنّاءة.

## مثال محسوب يدويًا

مجموعتا توصيل، إحداهما بنظام توجيه جديد:

- المسار الحالي: $n_1 = 30$، المتوسط 18.0 دقيقة، الانحراف المعياري 6.0.
- المسار الجديد: $n_2 = 30$، المتوسط 15.0 دقيقة، الانحراف المعياري 5.0.

**الخطوة 1 — الفرضيتان.** $H_0$: المتوسطان متساويان. $H_1$: المسار الجديد أسرع.

**الخطوة 2 — الأخطاء المعيارية.**

$$
\text{SE}_1 = \frac{6.0}{\sqrt{30}} \approx 1.10, \qquad \text{SE}_2 = \frac{5.0}{\sqrt{30}} \approx 0.91
$$

**الخطوة 3 — الخطأ المعياري للفرق.** في المجموعات المستقلة تُجمع التباينات:

$$
\text{SE}_{\text{diff}} = \sqrt{1.10^2 + 0.91^2} \approx \sqrt{1.21 + 0.83} \approx 1.43
$$

**الخطوة 4 — الإحصاءة.**

$$
t = \frac{18.0 - 15.0}{1.43} \approx 2.10
$$

**الخطوة 5 — القيمة الاحتمالية.** بنحو 58 درجة حرية، تتجاوز قيمة $t$ مقدار 2.10 في ذيل واحد باحتمال نحو 0.020. وبما أن $H_1$ تقول "أسرع"، فالقيمة الاحتمالية أحادية الطرف نحو 0.02؛ وثنائية الطرف نحو 0.04. على أي حال هي دون 0.05.

**الخطوة 6 — القرار، بصياغة صادقة.** "يرتبط المسار الجديد بمتوسط زمن توصيل أقصر بـ3 دقائق، $t \approx 2.10$، $p \approx 0.04$. وفترة الثقة للفرق نحو 0.1 إلى 5.9 دقيقة." لاحظ كم تقول الفترة أكثر من الحكم: فقد يكون الأثر تافهًا أو جوهريًا.

## ما لا تكونه القيمة الاحتمالية

- **ليست احتمال أن تكون $H_0$ صحيحة.** إنها محسوبة *بافتراض* أن $H_0$ صحيحة.
- **ليست احتمال أن تكون مخطئًا.** إنها عبارة عن البيانات، لا عن استنتاجك.
- **ليست مقياسًا لحجم الأثر.** أثر ضئيل مع عينة ضخمة ينتج قيمة احتمالية ضئيلة.
- **ليست رخصة للصيد.** اختبار فرضيات كثيرة والإبلاغ عن أصغر قيمة احتمالية يضمن نتائج إيجابية كاذبة؛ فمع 20 اختبارًا عند 0.05 تتوقع نتيجة "دالة" زائفة واحدة.

## الشيء نفسه في بايثون


```python
from scipy import stats

current = [17, 18, 19, 18, 16, 20]        # illustrative samples
new     = [14, 15, 16, 15, 13, 17]

res = stats.ttest_ind(new, current, alternative='less')
print(round(res.statistic, 2), round(res.pvalue, 4))
```

اطبع دائمًا الأثر وفترته إلى جانب الاختبار:


```python
import statistics as st

diff = st.mean(current) - st.mean(new)
se = (st.stdev(current)**2 / len(current) + st.stdev(new)**2 / len(new)) ** 0.5
print(round(diff, 2), round(diff - 2*se, 2), round(diff + 2*se, 2))
```

## رؤيته هندسيًا

إحصاءة t هي المسافة بين توزيعي معاينة مقيسةً بتشتتهما. في [مستكشف اختبار t من GeoGebra](https://www.geogebra.org/m/qzxjhqde) يمكنك إبعاد متوسطي المجموعتين أو تكبير العينتين ومراقبة انكماش التداخل — الصورة نفسها التي تهبط معها القيمة الاحتمالية.

## متى تستخدمه

- مقارنة محددة مسبقًا بين مجموعتين أو مقابل معيار مرجعي.
- التحقق من أن فرقًا مرصودًا أكبر مما ينتجه ضجيج المعاينة.
- كمدخل واحد لقرار يوازن أيضًا بين التكلفة وحجم الأثر والسياق.

## القيود

- **يجيب عن سؤال ضيق** — "هل الأثر صفر؟" — ولا يقول شيئًا عن الحجم أو الأهمية.
- **الافتراضات مهمة.** الاستقلال، وبيانات طبيعية تقريبًا أو عينات كبيرة؛ ومخالفتها تبطل القيمة الاحتمالية.
- **المقارنات المتعددة تنفخ النتائج الإيجابية الكاذبة** ما لم تُصحح.
- **يكافئ العينات الكبيرة بالدلالة** لأثر لا يهتم به أحد.
- **لا ينقذ تصميمًا متحيزًا.** اختبار على عينة متحيزة يعطي إجابة دقيقة عن سؤال خاطئ.

## أخطاء شائعة

- **اختيار الفرضية بعد رؤية البيانات**، مما يحوّل نتيجة استكشافية بهدوء إلى تأكيد زائف.
- **الإبلاغ عن p < 0.05 كأنه "إثبات"**؛ إنه دليل ضد نموذج صفري، لا إثبات لميكانيزم.
- **إخفاء حجم الأثر.** قيمة احتمالية بلا فرق في المتوسط ولا فترة غير قابلة للتفسير.
- **الخلط بين الاختبارات أحادية وثنائية الطرف**، مما يغيّر العتبة دون تغيير البيانات.
