## ماذا يعني

يلخّص **خط الانحدار** (regression line) كيف يتغير استجابة $y$ مع منبئ $x$. ويقلل خط المربعات الصغرى مجموع المسافات العمودية المربعة من النقاط إلى الخط:

$$
\hat{y} = b_0 + b_1 x, \qquad
b_1 = r\,\frac{s_y}{s_x}, \qquad
b_0 = \bar{y} - b_1 \bar{x}.
$$

الميل هو الارتباط معادًا تحجيمه إلى وحدات البيانات: اضرب $r$ في نسبة الانحرافين المعياريين فتحصل على "كم وحدة تتحرك $y$ لكل وحدة من $x$". ولهذا يتشارك $r$ و$b_1$ الإشارة دائمًا: إنهما العلاقة نفسها معبَّرًا عنها على مقياسين مختلفين.

## لماذا يضيف شيئًا

يقول الارتباط *إن* متغيرين يتحركان معًا؛ ويقول الانحدار *بكم*، بوحدات تستطيع التصرف بناءً عليها. "الارتباط 0.7" يصعب استخدامه؛ أما "كل ساعة مذاكرة إضافية ترتبط بـ6.3 نقطة إضافية" فجملة يمكن بناء قرار عليها.

ينتج الانحدار أيضًا البواقي — أجزاء $y$ التي يفشل الخط في تفسيرها — وفي تلك البواقي يحدث كل فحص للافتراضات.

## مثال محسوب يدويًا

الطلاب الخمسة مرة أخرى: الساعات $x = 1,2,3,4,5$؛ والدرجات $y = 50,55,65,70,80$.

من درس الارتباط: $\bar{x} = 3$، $\bar{y} = 64$، $s_x \approx 1.581$، $s_y \approx 11.94$، $r \approx 0.987$.

**الخطوة 1 — الميل.**

$$
b_1 = 0.987 \times \frac{11.94}{1.581} \approx 0.987 \times 7.55 \approx 7.45
$$

**الخطوة 2 — الثابت.**

$$
b_0 = 64 - 7.45 \times 3 = 64 - 22.35 = 41.65
$$

**الخطوة 3 — الخط.** $\hat{y} = 41.65 + 7.45x$.

**الخطوة 4 — اقرأه.** كل ساعة مذاكرة إضافية ترتبط بنحو 7.5 نقاط إضافية. والثابت، 41.65، هو الدرجة المتوقعة لصفر ساعات — متوسط شرطي، لا ادعاء بأن طالبًا لا يذاكر شيء يحصل على 41.65 بالمعنى السببي.

**الخطوة 5 — البواقي.** المتوقع مقابل الفعلي:

| $x$ | $y$ | $\hat{y}$ | الباقي |
|---|---|---|---|
| 1 | 50 | 49.1 | +0.9 |
| 2 | 55 | 56.6 | −1.6 |
| 3 | 65 | 64.0 | +1.0 |
| 4 | 70 | 71.4 | −1.4 |
| 5 | 80 | 78.9 | +1.1 |

مجموع البواقي صفر (حتى حدود التقريب)، وتتبادل الإشارة، وتبقى صغيرة — تلاؤم سليم لخط مستقيم.

## الافتراضات الأربعة

افحصها على البواقي، لا على مخطط الانتشار الخام:

1. **الخطية** — لا تُظهر البواقي انحناءً. النمط على شكل U يعني أن العلاقة تنحني.
2. **تساوي التشتت** — تتباعد البواقي (أو تضيق) كلما زاد $x$، وهذا عدم تجانس التباين؛ وحينها ستكون الأخطاء المعيارية من التركيب خاطئة.
3. **الاستقلال** — البواقي غير مرتبطة بترتيب الجمع أو المجموعة أو الزمن.
4. **الطبيعية التقريبية للبواقي** — مطلوبة للقيم الاحتمالية والفترات، وأقل أهمية لتقدير الميل نفسه.


```python
import numpy as np

x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([50, 55, 65, 70, 80], dtype=float)

b1, b0 = np.polyfit(x, y, 1)
print(round(b1, 2), round(b0, 2))         # 7.45 41.65

resid = y - (b0 + b1 * x)
print(np.round(resid, 2))                  # [ 0.9 -1.6  1.0 -1.4  1.1]
```

مع pandas وscipy تحصل على الميل نفسه زائد الخطأ المعياري الذي ستبلغ عنه:


```python
from scipy import stats
res = stats.linregress(x, y)
print(res.slope, res.intercept, res.rvalue, res.pvalue, res.stderr)
```

## رؤيته هندسيًا

خط المربعات الصغرى هو الخط الذي يقلل المساحة الكلية للمربعات المرسومة على كل باقٍ. اسحب نقطة في [مستكشف الانحدار من GeoGebra](https://www.geogebra.org/m/dcpcyc3t) وراقب الخط يتأرجح — ولاحظ أن نقطة واحدة بعيدة في نهاية المدى تعتله بقوة أكبر بكثير من نقطة قرب الوسط.

## متى تستخدمه

- تكميم علاقة بوحدات البيانات نفسها، بميل تستطيع اقتباسه.
- التعديل لمتغير ثانٍ، وهذا ما يفعله الانحدار المتعدد وما ينقذك من مفارقة سيمبسون.
- إنتاج تنبؤات داخل المدى المرصود، مع عدم اليقين الملحق.

## القيود

- **لا يستطيع الخط تمثيل منحنى.** إذا أظهرت البواقي انحناءً، فالنموذج خاطئ مهما كان $r$ مرتفعًا.
- **الارتباط لا يزال ليس سببية.** ميل الانحدار يرث كل العوامل المربكة التي كان عليها الارتباط.
- **النقاط المؤثرة قد تحدد الميل.** نقطة واحدة عند $x$ متطرف قد تحرك الخط بشكل هائل؛ فافحص بإعادة التركيب دونها.
- **الاستقراء الخارجي خطير** بعد المدى المركّب، كما يُظهر مثال الـ10 ساعات.
- **المدى المقيد يضعف الميل.** ركّب فقط بين ذوي الأداء العالي فيتسطح الميل.

## أخطاء شائعة

- **الإبلاغ عن $r$ بينما ركّبت ميلًا** (أو العكس) دون بيان أيها.
- **قراءة الثابت كخط أساس واقعي** بينما $x = 0$ خارج البيانات.
- **تجاهل مخطط البواقي** لأن $r$ كان مرتفعًا.
- **وصف الميل بالسببي** بينما كان التصميم رصديًا.

## 🤔 أسئلة سقراطية

- لماذا يمر خط المربعات الصغرى بالنقطة $(\bar{x}, \bar{y})$؟
- إذا كان $r = 0$، فما ميل المربعات الصغرى، وماذا يتوقع الخط لكل $x$؟
- لماذا يكون مجموع البواقي صفرًا دائمًا في نموذج له ثابت؟
- ماذا يعني مخطط بواقٍ مروحي الشكل بخصوص فترة الثقة للميل؟
